# Datetime

## Understand the Concept

In Pandas, datetime is used to work with date and time values in a structured way.

It allows us to store, compare, and perform operations on dates and times.

### Why do we use it?

- To work with date and time data.
- To compare dates.
- To perform date and time calculations.
- To analyze time-based data.

### AI/ML Use

In AI/ML projects, datasets often contain date and time information.

Datetime is useful for analyzing time-based patterns such as sales trends, customer activity, sensor readings, and machine usage.

## Demonstrate the Concept

### Real-World Example

Suppose a company records the date and time when a customer places an order.

Instead of treating the date as normal text, we can convert it into a datetime value.

This allows us to compare orders, calculate time differences, and analyze sales based on dates.

### AI/ML Example

Suppose a machine learning dataset contains the date and time when a customer made a purchase.

We can convert this information into datetime format and later use it to identify patterns such as daily, weekly, or monthly purchasing behavior.

```python
import pandas as pd

# Create a DataFrame containing date and time values
data = pd.DataFrame({
    "Order": [1, 2, 3],
    "Order_Date": [
        "2026-10-01 10:30:00",
        "2026-10-02 14:45:00",
        "2026-10-03 09:15:00"
    ]
})

# Display the original data
print("Original Data:")
print(data)

# Convert the Order_Date column into datetime format
# pd.to_datetime() converts text dates into datetime values.
data["Order_Date"] = pd.to_datetime(data["Order_Date"])

# Display the converted data
print("\nData after converting to datetime:")
print(data)

# Check the data type of the Order_Date column
print("\nData Type:")
print(data["Order_Date"].dtype)

# Expected output:
# The Order_Date column will have datetime64[ns] as its data type.

# AI/ML / Data Analytics Use:
# Datetime values can be used to analyze time-based patterns,
# calculate time differences, and create useful date features.
```

# Timestamp

## Understand the Concept

In Pandas, `Timestamp` represents a specific date and time.

It is similar to Python's `datetime` object and is commonly used when working with individual date and time values in Pandas.

### Why do we use it?

- To represent a specific date and time.
- To compare date and time values.
- To perform date and time calculations.
- To work with time-based data in Pandas.

### AI/ML Use

In AI/ML projects, `Timestamp` can be used to represent events such as customer transactions, sensor readings, system logs, or machine activity at a specific point in time.

## Demonstrate the Concept

### Real-World Example

Suppose a company records the exact time when a customer places an order.

For example, an order may be placed on October 5, 2026 at 10:30 AM.

We can represent this exact date and time using a Pandas `Timestamp`.

### AI/ML Example

Suppose a sensor records the temperature of a machine at a specific time.

We can use `Timestamp` to identify exactly when each sensor reading was collected.

This helps in analyzing time-based patterns and detecting changes in machine behavior.

```python
import pandas as pd

# Create a Pandas Timestamp
order_time = pd.Timestamp("2026-10-05 10:30:00")

# Display the Timestamp
print("Order Time:")
print(order_time)

# Display individual date and time components
print("\nYear:", order_time.year)
print("Month:", order_time.month)
print("Day:", order_time.day)
print("Hour:", order_time.hour)
print("Minute:", order_time.minute)

# Expected output:
# The program displays the complete timestamp and
# its year, month, day, hour, and minute.

# AI/ML / Data Analytics Use:
# Timestamp is useful for identifying the exact time
# when events, transactions, or sensor readings occur.
```

# Date Parsing

## Understand the Concept

Date parsing means converting date values from different formats into a standard datetime format.

In Pandas, `pd.to_datetime()` is commonly used to parse date strings and convert them into datetime values.

### Why do we use it?

- To convert text dates into datetime format.
- To handle different date formats.
- To make date-based calculations easier.
- To prepare date data for analysis.

### AI/ML Use

In AI/ML projects, datasets may contain dates in different formats.

Date parsing helps convert these values into a consistent datetime format before performing data analysis, feature engineering, or model training.

## Demonstrate the Concept

### Real-World Example

Suppose a company receives customer data from different systems.

One system stores dates as `2026-10-01`, while another stores dates as `10/02/2026`.

We can use date parsing to convert these date values into a standard datetime format.

### AI/ML Example

Suppose a sales dataset contains transaction dates in different formats.

We can parse these dates into a consistent datetime format and then use them to analyze daily, monthly, or yearly sales patterns.

```python
import pandas as pd

# Create a DataFrame with dates in different formats
data = pd.DataFrame({
    "Date": [
        "2026-10-01",
        "10/02/2026",
        "2026-10-03"
    ]
})

# Display the original data
print("Original Data:")
print(data)

# Convert dates with different formats into datetime
# format="mixed" allows Pandas to identify the format
# of each date individually.
data["Date"] = pd.to_datetime(data["Date"], format="mixed")

# Display the converted data
print("\nAfter Date Parsing:")
print(data)

# Check the data type
print("\nData Type:")
print(data["Date"].dtype)

# Expected output:
# The Date column will be converted into datetime64[ns].

# AI/ML / Data Analytics Use:
# Date parsing helps prepare date information for
# time-based analysis and feature engineering.
```

# Timedelta

## Understand the Concept

In Pandas, `Timedelta` represents a duration or difference between two dates or times.

It can be used to calculate how much time has passed between two datetime values.

### Why do we use it?

- To calculate the difference between dates.
- To add or subtract days and time.
- To calculate durations.
- To measure the time taken between events.

### AI/ML Use

In AI/ML projects, `Timedelta` can be used to calculate customer response time, delivery duration, machine operating time, or the time between two events.

## Demonstrate the Concept

### Real-World Example

Suppose a customer places an order on October 1 and the order is delivered on October 4.

We can use `Timedelta` to calculate how many days it took to deliver the order.

### AI/ML Example

Suppose a company wants to predict delivery delays.

We can calculate the number of days between the order date and delivery date.

This duration can be used as a feature for data analysis or machine learning.

```python
import pandas as pd

# Create order and delivery dates
order_date = pd.Timestamp("2026-10-01")
delivery_date = pd.Timestamp("2026-10-04")

# Calculate the difference between the two dates
# The subtraction of two Timestamp values returns a Timedelta.
delivery_time = delivery_date - order_date

# Display the dates
print("Order Date:", order_date)
print("Delivery Date:", delivery_date)

# Display the time difference
print("\nDelivery Duration:")
print(delivery_time)

# Get the number of days from the Timedelta
print("\nNumber of Days:", delivery_time.days)

# Expected output:
# Delivery Duration: 3 days
# Number of Days: 3

# AI/ML / Data Analytics Use:
# Timedelta can be used to calculate durations such as
# delivery time, customer response time, or machine usage time.
```

# Time Zones

## Understand the Concept

A time zone represents the local time of a particular region of the world.

In Pandas, time zones can be used to represent and convert datetime values between different locations.

### Why do we use it?

- To work with dates and times from different regions.
- To convert time between time zones.
- To avoid confusion when working with global data.
- To correctly analyze events that happen in different locations.

### AI/ML Use

In AI/ML projects, data may be collected from users, servers, sensors, or businesses located in different countries.

Time zones help standardize and correctly analyze this time-based data.

## Demonstrate the Concept

### Real-World Example

Suppose a company has customers in India and the United States.

The same event may occur at different local times in these countries.

We can use time zones to represent and convert the event time correctly for each location.

### AI/ML Example

Suppose an online application records user activity from different countries.

Time zone information can help identify the correct local time of each activity.

This can be useful for analyzing user behavior and creating time-based features for machine learning.

```python
import pandas as pd

# Create a datetime value in UTC
utc_time = pd.Timestamp("2026-10-08 10:00:00", tz="UTC")

# Convert the UTC time to India Standard Time
india_time = utc_time.tz_convert("Asia/Kolkata")

# Convert the UTC time to US Eastern Time
us_time = utc_time.tz_convert("America/New_York")

# Display the original UTC time
print("UTC Time:")
print(utc_time)

# Display the converted times
print("\nIndia Time:")
print(india_time)

print("\nUS Eastern Time:")
print(us_time)

# Expected output:
# The same moment will be displayed with different
# local times and time zone information.

# AI/ML / Data Analytics Use:
# Time zones are useful when analyzing global user activity,
# transactions, server logs, and sensor data.
```

# Extracting Date Features

## Understand the Concept

Extracting date features means getting useful information such as year, month, day, day of the week, and hour from a datetime value.

In Pandas, we can use the `.dt` accessor to extract different date and time components.

### Why do we use it?

- To extract useful information from dates.
- To analyze data based on year, month, day, or time.
- To create new features from datetime data.
- To identify patterns in time-based data.

### AI/ML Use

In AI/ML projects, date features can be used as input features for machine learning models.

For example, we can extract the month, day of the week, or hour from a transaction date to help a model identify customer behavior and seasonal patterns.

## Demonstrate the Concept

### Real-World Example

Suppose a company has a sales date for every customer order.

Instead of using only the complete date, we can extract the year, month, day, and day of the week.

This helps the company understand which months or days have higher sales.

### AI/ML Example

Suppose an online shopping company wants to predict the number of orders.

We can extract features such as month, day of the week, and hour from the order date.

These features can help a machine learning model learn purchasing patterns and make better predictions.

```python
import pandas as pd

# Create a DataFrame containing order dates
data = pd.DataFrame({
    "Order_Date": [
        "2026-10-05 10:30:00",
        "2026-11-15 14:45:00",
        "2026-12-20 18:15:00"
    ]
})

# Convert the Order_Date column into datetime format
data["Order_Date"] = pd.to_datetime(data["Order_Date"])

# Extract useful date features using the .dt accessor
data["Year"] = data["Order_Date"].dt.year
data["Month"] = data["Order_Date"].dt.month
data["Day"] = data["Order_Date"].dt.day
data["Day_of_Week"] = data["Order_Date"].dt.day_name()
data["Hour"] = data["Order_Date"].dt.hour

# Display the DataFrame
print("Data with Extracted Date Features:")
print(data)

# Expected output:
# New columns will contain the year, month, day,
# day of the week, and hour from the Order_Date.

# AI/ML / Data Analytics Use:
# Extracted date features can be used for data analysis,
# visualization, feature engineering, and machine learning models.
```
