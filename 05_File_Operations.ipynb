# CSV Files

## Understand the Concept

CSV stands for **Comma-Separated Values**.

A CSV file stores data in a simple table-like format where values are separated by commas.

Pandas provides:
- `pd.read_csv()` – to read data from a CSV file
- `df.to_csv()` – to write DataFrame data into a CSV file

CSV is one of the most commonly used file formats for storing and sharing structured data.

In AI/ML, CSV files are commonly used to **store datasets and load them for data preprocessing, analysis, and model training**.

## Demonstrate the Concept

### Real-World Example

Suppose a company stores customer information in a CSV file:

Name,Age,Purchase
Arun,25,5000
Ravi,30,7000
Priya,28,6000

We can use Pandas to read this CSV file into a DataFrame.

After processing the data, we can also save the DataFrame back into a CSV file.

### AI/ML Example

Suppose we receive a customer dataset in CSV format for a machine learning project.

We can use `pd.read_csv()` to load the dataset, clean the data, select features, and prepare it for model training.

After preprocessing, we can use `to_csv()` to save the processed dataset.

So, CSV files are useful for **storing, sharing, reading, and writing structured datasets**.

```python
import pandas as pd

# Create sample data
data = {
    "Name": ["Arun", "Ravi", "Priya"],
    "Age": [25, 30, 28],
    "Purchase": [5000, 7000, 6000]
}

df = pd.DataFrame(data)

# Write the DataFrame to a CSV file
df.to_csv("customers.csv", index=False)

# Read the CSV file
read_data = pd.read_csv("customers.csv")

print(read_data)

# Why to_csv() is used:
# It saves the DataFrame as a CSV file.
# index=False prevents the DataFrame index from being saved.

# Why read_csv() is used:
# It reads the CSV file and converts it into a DataFrame.

# Expected Output:
#     Name  Age  Purchase
# 0   Arun   25      5000
# 1   Ravi   30      7000
# 2  Priya   28      6000

# AI/ML/Data Analytics Usage:
# CSV files are commonly used to store datasets
# that are later loaded for preprocessing and analysis.
```

# Excel Files

## Understand the Concept

Excel files are commonly used to store structured data in the form of **rows and columns**.

Pandas provides:
- `pd.read_excel()` – to read data from an Excel file
- `df.to_excel()` – to write DataFrame data into an Excel file

Excel files usually have the `.xlsx` extension.

In AI/ML and Data Analytics, Excel files can be used to **store, share, and analyze structured datasets** before performing further processing.

## Demonstrate the Concept

### Real-World Example

Suppose a company maintains employee information in an Excel file:

Name | Age | Salary
Arun | 25 | 30000
Ravi | 30 | 40000
Priya | 28 | 35000

Using Pandas, we can read this Excel file into a DataFrame.

After processing the data, we can also save the DataFrame back into an Excel file.

### AI/ML Example

Suppose a team receives customer data in an Excel file.

We can use `pd.read_excel()` to load the data, inspect and clean it, and prepare the required features for analysis or machine learning.

After processing, we can use `to_excel()` to save the updated data.

So, Pandas makes it easy to **read and write Excel data programmatically**.

```python
import pandas as pd

# Create sample employee data
data = {
    "Name": ["Arun", "Ravi", "Priya"],
    "Age": [25, 30, 28],
    "Salary": [30000, 40000, 35000]
}

df = pd.DataFrame(data)

# Write the DataFrame to an Excel file
df.to_excel("employees.xlsx", index=False)

# Read the Excel file
read_data = pd.read_excel("employees.xlsx")

print(read_data)

# Why to_excel() is used:
# It saves the DataFrame as an Excel file.
# index=False prevents the DataFrame index from being saved.

# Why read_excel() is used:
# It reads the Excel file and converts it into a DataFrame.

# Expected Output:
#     Name  Age  Salary
# 0   Arun   25   30000
# 1   Ravi   30   40000
# 2  Priya   28   35000

# AI/ML/Data Analytics Usage:
# Excel files can be used to store and share structured
# datasets before analysis and machine learning preprocessing.
```

# JSON Files

## Understand the Concept

JSON stands for **JavaScript Object Notation**.

It is a common format used to store and exchange structured data.

JSON represents data using **key-value pairs**, similar to a Python dictionary.

Pandas provides:
- `pd.read_json()` – to read JSON data
- `df.to_json()` – to write DataFrame data into a JSON file

In AI/ML and Data Analytics, JSON files are commonly used when **data comes from APIs, web applications, or other software systems**.

## Demonstrate the Concept

### Real-World Example

Suppose a web application provides customer information in JSON format:

{
    "Name": "Arun",
    "Age": 25,
    "Purchase": 5000
}

Multiple customer records can be stored in a JSON file.

Pandas can read this JSON data into a DataFrame, making it easier to analyze and process.

### AI/ML Example

Suppose an application provides customer or product information through an API in JSON format.

We can load the JSON data into Pandas, clean and analyze it, and then use the required data for machine learning or data analysis.

So, JSON is useful for **storing and exchanging structured data between applications**.

```python
import pandas as pd

# Create sample customer data
data = {
    "Name": ["Arun", "Ravi", "Priya"],
    "Age": [25, 30, 28],
    "Purchase": [5000, 7000, 6000]
}

df = pd.DataFrame(data)

# Write the DataFrame to a JSON file
df.to_json("customers.json", orient="records", indent=4)

# Read the JSON file
read_data = pd.read_json("customers.json")

print(read_data)

# Why to_json() is used:
# It saves the DataFrame in JSON format.
# orient="records" stores each row as a separate JSON object.

# Why read_json() is used:
# It reads JSON data and converts it into a DataFrame.

# Expected Output:
#     Name  Age  Purchase
# 0   Arun   25      5000
# 1   Ravi   30      7000
# 2  Priya   28      6000

# AI/ML/Data Analytics Usage:
# JSON is commonly used when collecting data from APIs
# and applications before processing it with Pandas.
```

# Parquet Files

## Understand the Concept

Parquet is a **column-based file format** commonly used for storing large datasets efficiently.

Unlike CSV files, Parquet stores data by columns, which makes it efficient for reading only the columns we need.

Parquet files are also **compressed**, so they usually require less storage space than many traditional formats.

Pandas provides:
- `pd.read_parquet()` – to read Parquet files
- `df.to_parquet()` – to write DataFrame data into a Parquet file

In AI/ML and Data Analytics, Parquet is useful for **large datasets and efficient data processing**.

## Demonstrate the Concept

### Real-World Example

Suppose a company has millions of customer records with columns such as:

Customer_ID | Age | Location | Purchase

If we only need the `Age` and `Purchase` columns, a column-based format like Parquet can efficiently read the required columns instead of processing the complete dataset.

### AI/ML Example

Suppose an AI/ML project has a large dataset containing millions of records.

Parquet can help store the processed dataset efficiently because:
- It supports compression
- It stores data by columns
- It can efficiently read selected columns

So, Parquet is useful when working with **large datasets and performance-sensitive data processing**.

```python
import pandas as pd

data = {
    "Name": ["Arun", "Ravi", "Priya"],
    "Age": [25, 30, 28],
    "Purchase": [5000, 7000, 6000]
}

df = pd.DataFrame(data)

df.to_parquet("customers.parquet", index=False)

print("Parquet file created successfully!")

read_data = pd.read_parquet("customers.parquet")

print(read_data)
```

# Pickle Files

## Understand the Concept

Pickle is a Python module used to save Python objects into a file and load them back later.

It can store objects such as:
- Lists
- Dictionaries
- DataFrames
- Machine Learning models
- Other Python objects

The file is usually saved with a `.pkl` or `.pickle` extension.

Pickle is useful when we want to save Python data in its original structure and reuse it later without creating the object again.

## Demonstrate the Concept

### Real-World Example

Suppose we have customer data stored in a Python DataFrame.

Instead of creating the DataFrame again every time, we can save it as a Pickle file. Later, we can load the file and directly use the DataFrame.

### AI/ML Example

In Machine Learning, Pickle can be used to save a trained model.

After training a model, we can save it as a `.pkl` file and load it later for making predictions without training the model again.

```python
import pandas as pd

# Create a simple DataFrame
# We use a DataFrame to represent customer data.
data = {
    "Name": ["Arun", "Priya", "Kumar"],
    "Age": [25, 28, 30],
    "Purchase": [500, 750, 1000]
}

df = pd.DataFrame(data)

# Save the DataFrame as a Pickle file
# to_pickle() stores the DataFrame in Python's pickle format.
df.to_pickle("customer_data.pkl")

# Load the Pickle file
# read_pickle() reads the saved Python object back into memory.
loaded_df = pd.read_pickle("customer_data.pkl")

# Display the loaded DataFrame
print(loaded_df)

# Expected Output:
#     Name  Age  Purchase
# 0   Arun   25       500
# 1  Priya   28       750
# 2  Kumar   30      1000

# AI/ML / Data Analytics Use:
# Pickle can be used to save trained ML models,
# preprocessing objects, or DataFrames and reuse them later.
```

# SQL Tables

## Understand the Concept

Pandas can read data directly from SQL databases and can also write DataFrames into SQL tables.

A SQL table stores data in rows and columns, similar to a Pandas DataFrame.

Pandas provides functions such as:
- `pd.read_sql()` – reads data from a SQL query or table.
- `df.to_sql()` – writes a DataFrame into a SQL table.

This is useful when working with large business and application databases.

## Demonstrate the Concept

### Real-World Example

A company may store customer information in a SQL database.

For example, a `customers` table may contain:

| Name | Age | Purchase |
|------|-----|----------|
| Arun | 25  | 500      |
| Priya | 28 | 750      |
| Kumar | 30 | 1000     |

We can load this SQL table into a Pandas DataFrame for data analysis.

### AI/ML Example

In AI/ML projects, data is often stored in databases.

We can retrieve the required data from a SQL table using Pandas, perform data cleaning and analysis, and then use the prepared data for Machine Learning.

```python
import pandas as pd
import sqlite3

# Create a connection to a SQLite database.
# SQLite is a simple database that does not require a separate server.
connection = sqlite3.connect("customer_database.db")

# Create a DataFrame with customer data.
data = {
    "Name": ["Arun", "Priya", "Kumar"],
    "Age": [25, 28, 30],
    "Purchase": [500, 750, 1000]
}

df = pd.DataFrame(data)

# Write the DataFrame into a SQL table.
# to_sql() creates a table and stores the DataFrame data in it.
df.to_sql("customers", connection, if_exists="replace", index=False)

# Read the SQL table into a Pandas DataFrame.
# read_sql() executes the SQL query and returns the result as a DataFrame.
loaded_df = pd.read_sql("SELECT * FROM customers", connection)

# Display the data retrieved from SQL.
print(loaded_df)

# Expected Output:
#     Name  Age  Purchase
# 0   Arun   25       500
# 1  Priya   28       750
# 2  Kumar   30      1000

# AI/ML / Data Analytics Use:
# SQL is commonly used to retrieve data from databases,
# while Pandas is used for further cleaning, analysis,
# visualization, and Machine Learning preparation.

# Close the database connection.
connection.close()
```

# Reading Multiple Files

## Understand the Concept

When data is stored in multiple files, Pandas can be used to read and combine those files into one DataFrame.

For example, a company may store sales data separately for:
- January
- February
- March

Instead of reading each file manually, we can use Python to find multiple files and process them together.

This is useful when working with datasets collected from different sources or different time periods.

## Demonstrate the Concept

### Real-World Example

Suppose a company has three CSV files:

- `sales_january.csv`
- `sales_february.csv`
- `sales_march.csv`

Each file contains sales information.

We can read all the CSV files and combine their data into one DataFrame for analysis.

### AI/ML Example

In an AI/ML project, data may be collected and stored in multiple files.

For example, customer data from different months can be combined into one dataset before performing data cleaning, analysis, and Machine Learning.

```python
import pandas as pd
import glob

# Create sample CSV files for three months.
# In a real project, these files would already exist.
january = pd.DataFrame({
    "Name": ["Arun", "Priya"],
    "Purchase": [500, 700]
})

february = pd.DataFrame({
    "Name": ["Kumar", "Divya"],
    "Purchase": [800, 600]
})

march = pd.DataFrame({
    "Name": ["Ravi", "Anu"],
    "Purchase": [900, 750]
})

january.to_csv("sales_january.csv", index=False)
february.to_csv("sales_february.csv", index=False)
march.to_csv("sales_march.csv", index=False)

# Find all CSV files that start with "sales_".
# glob() helps us find multiple files using a pattern.
files = glob.glob("sales_*.csv")

# Read each CSV file into a DataFrame.
dataframes = [pd.read_csv(file) for file in files]

# Combine all DataFrames into one DataFrame.
# concat() combines data row by row.
combined_df = pd.concat(dataframes, ignore_index=True)

# Display the combined data.
print(combined_df)

# Expected Output:
#     Name  Purchase
# 0   Arun       500
# 1  Priya       700
# 2  Kumar       800
# 3  Divya       600
# 4   Ravi       900
# 5    Anu       750

# AI/ML / Data Analytics Use:
# Reading multiple files is useful when datasets are split
# by month, region, department, or other categories.
# Combining them allows us to perform analysis or prepare
# one complete dataset for Machine Learning.
```

# File Path Handling

## Understand the Concept

File path handling means working with the location of files and folders in a safe and organized way.

Python provides the `pathlib` module to work with file paths.

Instead of manually writing file paths as strings, we can use `Path` from `pathlib`.

It helps us:
- Create file paths
- Check whether a file exists
- Find files in a folder
- Work with folders
- Make file handling easier across different operating systems

## Demonstrate the Concept

### Real-World Example

Suppose our project has a folder called `data` containing several CSV files.

Instead of manually writing:

`data/customer.csv`

we can use `Path` to create and manage the file path.

This makes the code cleaner and easier to maintain.

### AI/ML Example

In an AI/ML project, we commonly work with folders such as:

- `data/` – datasets
- `models/` – trained models
- `output/` – prediction results
- `logs/` – log files

Using `pathlib` helps us manage these files and folders programmatically.

```python
from pathlib import Path

# Create a Path object for the data folder.
# Path() is used to create and manage file paths.
data_folder = Path("data")

# Create the folder if it does not already exist.
# mkdir() creates a directory.
data_folder.mkdir(exist_ok=True)

# Create a file path inside the data folder.
# The / operator joins folder and file names safely.
file_path = data_folder / "customers.csv"

# Display the file path.
print("File Path:", file_path)

# Check whether the file exists.
# exists() returns True if the file is present.
print("File Exists:", file_path.exists())

# Display the folder name.
print("Folder Name:", file_path.parent)

# Display the file name.
print("File Name:", file_path.name)

# Expected Output:
# File Path: data\customers.csv
# File Exists: False
# Folder Name: data
# File Name: customers.csv

# AI/ML / Data Analytics Use:
# pathlib is useful for managing dataset files,
# trained model files, output files, and project folders.
# It makes file handling cleaner and more reliable.
```
