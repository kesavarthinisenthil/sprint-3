# 1. What is NumPy?

NumPy stands for Numerical Python. It is a Python library used for working with numerical data and performing mathematical operations efficiently.

The main data structure provided by NumPy is called an `ndarray` (N-dimensional array). It allows us to store and process multiple numerical values together.

NumPy is useful because it provides fast array operations, mathematical functions, and support for multidimensional data.

It is widely used in Data Science, Artificial Intelligence, Machine Learning, and scientific computing.

# Real-World / AI-ML Example

Suppose we have the marks of five students:

[80, 75, 90, 85, 95]

If we want to calculate the average mark or add 5 marks to every student's mark, NumPy allows us to perform these operations directly on the entire collection of values.

In AI/ML, we frequently work with numerical data such as age, salary, temperature, prices, and model features. NumPy helps us store and perform calculations on this numerical data efficiently.

```python
import numpy as np

# Create a NumPy array containing student marks
marks = np.array([80, 75, 90, 85, 95])

# Calculate the average using NumPy
average = np.mean(marks)

# Add 5 marks to every student
updated_marks = marks + 5

print("Original Marks:", marks)
print("Average Marks:", average)
print("Updated Marks:", updated_marks)

# Explanation:
# np.array() creates a NumPy ndarray from the given list.
# np.mean() calculates the average of all values.
# marks + 5 adds 5 to every element of the array.
#
# Expected Output:
# Original Marks: [80 75 90 85 95]
# Average Marks: 85.0
# Updated Marks: [85 80 95 90 100]
#
# AI/ML Use:
# Similar array operations are used to process numerical
# features before performing data analysis or training ML models.
```

# 2. Why NumPy?

NumPy is used to work with numerical data efficiently.

Python lists can store multiple values, but NumPy arrays are designed specifically for numerical operations. NumPy allows us to perform calculations on many values at the same time without writing a separate loop for each value.

NumPy also provides useful features such as multidimensional arrays, mathematical functions, vectorized operations, and broadcasting.

Because of these features, NumPy is widely used in Data Analysis, Machine Learning, Artificial Intelligence, and scientific computing.

# Simple and AI/ML Example

Suppose we have the prices of four products:

[100, 200, 300, 400]

If we want to increase every price by 10%, we need to perform the same calculation on every value.

With NumPy, we can apply the calculation to the complete array at once instead of manually processing each value.

```python
import numpy as np

# Create a NumPy array containing product prices
prices = np.array([100, 200, 300, 400])

# Increase every price by 10%
updated_prices = prices * 1.10

print("Original Prices:", prices)
print("Updated Prices:", updated_prices)

# Explanation:
# np.array() creates a NumPy array.
# Multiplying the array by 1.10 increases every value by 10%.
# NumPy applies the operation to all elements at once.
#
# Expected Output:
# Original Prices: [100 200 300 400]
# Updated Prices: [110. 220. 330. 440.]
#
# AI/ML Use:
# Similar operations are used when transforming and scaling
# numerical features in Machine Learning datasets.
```

# 3. NumPy Arrays (`ndarray`)

A NumPy array is a data structure used to store multiple values in an organized way.

The main array object in NumPy is called `ndarray`, which means N-dimensional array.

Unlike a normal Python list, a NumPy array is designed for numerical operations and can efficiently perform calculations on multiple values.

A NumPy array can have one or more dimensions, such as 1-D, 2-D, or 3-D.

# Real-World / AI-ML Example

Consider the marks of three students:

[80, 75, 90]

These values can be stored in a 1-D NumPy array.

If we have marks for three students in three subjects:

[
    [80, 75, 90],
    [85, 70, 95],
    [90, 88, 92]
]

we can store them in a 2-D NumPy array, where rows represent students and columns represent subjects.

In AI/ML, datasets are commonly represented as 2-D arrays, where rows represent observations and columns represent features.

```python
import numpy as np

# Create a 1-D NumPy array
marks = np.array([80, 75, 90])

# Create a 2-D NumPy array
student_data = np.array([
    [80, 75, 90],
    [85, 70, 95],
    [90, 88, 92]
])

print("1-D Array:", marks)
print("Type:", type(marks))

print("\n2-D Array:")
print(student_data)

# Explanation:
# np.array() converts a Python list into a NumPy ndarray.
# The first array contains values in one dimension.
# The second array contains rows and columns, so it is 2-D.
# type() confirms that the object is a NumPy ndarray.
#
# Expected Output:
# 1-D Array: [80 75 90]
# Type: <class 'numpy.ndarray'>
#
# 2-D Array:
# [[80 75 90]
#  [85 70 95]
#  [90 88 92]]
#
# AI/ML Use:
# 2-D arrays are commonly used to represent datasets,
# where rows are samples and columns are features.
```

# 4. Creating NumPy Arrays

NumPy provides different functions to create arrays depending on the type of data we need.

Some commonly used functions are:

- `np.array()` – creates an array from a list or other sequence.
- `np.zeros()` – creates an array filled with zeros.
- `np.ones()` – creates an array filled with ones.
- `np.arange()` – creates an array with values within a given range.
- `np.linspace()` – creates an array with a specified number of evenly spaced values.

These functions make it easy to create and initialize numerical data for calculations and analysis.

# Simple / AI-ML Example

Suppose we are preparing data for a Machine Learning experiment.

We may need:

- An array containing existing data → `np.array()`
- An array initialized with zeros → `np.zeros()`
- An array initialized with ones → `np.ones()`
- A sequence of numbers → `np.arange()`
- Evenly spaced values for testing or numerical calculations → `np.linspace()`

For example, `np.zeros()` can be useful for initializing an array before storing calculated values, while `np.arange()` can be useful for creating sample indexes or numerical sequences.

```python
import numpy as np

# Create an array from a Python list
array1 = np.array([10, 20, 30, 40, 50])

# Create an array containing five zeros
array2 = np.zeros(5)

# Create an array containing five ones
array3 = np.ones(5)

# Create values from 1 up to, but not including, 6
array4 = np.arange(1, 6)

# Create five evenly spaced values between 0 and 10
array5 = np.linspace(0, 10, 5)

print("np.array():", array1)
print("np.zeros():", array2)
print("np.ones():", array3)
print("np.arange():", array4)
print("np.linspace():", array5)

# Explanation:
# np.array() creates an array from the given list.
# np.zeros(5) creates five zeros.
# np.ones(5) creates five ones.
# np.arange(1, 6) creates values from 1 to 5.
# np.linspace(0, 10, 5) creates 5 evenly spaced values
# between 0 and 10.
#
# Expected Output:
# np.array(): [10 20 30 40 50]
# np.zeros(): [0. 0. 0. 0. 0.]
# np.ones(): [1. 1. 1. 1. 1.]
# np.arange(): [1 2 3 4 5]
# np.linspace(): [ 0.   2.5  5.   7.5 10. ]
#
# AI/ML Use:
# These functions are useful for creating datasets, initializing
# arrays, generating test data, and preparing numerical values
# for Machine Learning and Data Analysis.
```

# 5. Dimensions

The dimension of a NumPy array tells us how many levels of structure the array has.

A 1-D array contains a single sequence of values.

A 2-D array contains rows and columns, similar to a table.

A 3-D array contains multiple 2-D arrays, creating another level of structure.

NumPy provides the `ndim` attribute to find the number of dimensions in an array.

# Real-World / AI-ML Example

A list of student marks can be represented using a 1-D array:

[80, 90, 75]

Marks of multiple students across multiple subjects can be represented using a 2-D array:

[
    [80, 90, 75],
    [85, 88, 92],
    [70, 78, 80]
]

In AI/ML, a dataset is commonly represented as a 2-D structure where rows represent samples and columns represent features.

Images can require higher-dimensional arrays because they contain height, width, and color-channel information.

```python
import numpy as np

# Create a 1-D array
array_1d = np.array([10, 20, 30, 40])

# Create a 2-D array
array_2d = np.array([
    [10, 20],
    [30, 40]
])

# Create a 3-D array
array_3d = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print("1-D array:", array_1d)
print("Number of dimensions:", array_1d.ndim)

print("\n2-D array:")
print(array_2d)
print("Number of dimensions:", array_2d.ndim)

print("\n3-D array:")
print(array_3d)
print("Number of dimensions:", array_3d.ndim)

# Explanation:
# The ndim attribute returns the number of dimensions.
# array_1d has one dimension.
# array_2d has two dimensions (rows and columns).
# array_3d has three dimensions.
#
# Expected Output:
# 1-D array: [10 20 30 40]
# Number of dimensions: 1
#
# 2-D array:
# [[10 20]
#  [30 40]]
# Number of dimensions: 2
#
# 3-D array:
# [[[1 2]
#   [3 4]]
#  [[5 6]
#   [7 8]]]
# Number of dimensions: 3
#
# AI/ML Use:
# Understanding dimensions is important when working with
# datasets, matrices, images, and other multidimensional data.
```

# 6. Shape

The shape of a NumPy array tells us how many elements are present along each dimension.

The `shape` attribute returns the size of each dimension as a tuple.

For a 1-D array, the shape tells us the number of elements.

For a 2-D array, the shape tells us the number of rows and columns.

Understanding the shape of data is important because many NumPy and Machine Learning operations require arrays to have compatible shapes.

# Simple / AI-ML Example

Suppose we have a dataset containing information about 3 students and 4 subjects:

[
    [80, 75, 90, 85],
    [70, 88, 82, 91],
    [95, 90, 87, 93]
]

This data has 3 rows and 4 columns.

Therefore, its shape is `(3, 4)`.

In AI/ML, the shape of a dataset helps us understand how many samples and features are present.

For example, a shape of `(100, 5)` usually means 100 samples and 5 features.

```python
import numpy as np

# Create a 1-D array
marks = np.array([80, 75, 90, 85])

# Create a 2-D array with 3 rows and 4 columns
student_data = np.array([
    [80, 75, 90, 85],
    [70, 88, 82, 91],
    [95, 90, 87, 93]
])

print("1-D Array:", marks)
print("1-D Shape:", marks.shape)

print("\n2-D Array:")
print(student_data)
print("2-D Shape:", student_data.shape)

# Explanation:
# The shape attribute returns the size of each dimension.
# marks has 4 elements, so its shape is (4,).
# student_data has 3 rows and 4 columns, so its shape is (3, 4).
#
# Expected Output:
# 1-D Array: [80 75 90 85]
# 1-D Shape: (4,)
#
# 2-D Array:
# [[80 75 90 85]
#  [70 88 82 91]
#  [95 90 87 93]]
# 2-D Shape: (3, 4)
#
# AI/ML Use:
# Shape is used to understand the structure of datasets and
# to make sure input data has the correct dimensions before
# passing it to Machine Learning algorithms.
```

# 7. Size

The `size` attribute of a NumPy array tells us the total number of elements present in the array.

It counts every element across all dimensions.

For example, if a 2-D array has 3 rows and 4 columns, its shape is `(3, 4)` and its size is `12` because there are 3 × 4 = 12 elements.

The `size` attribute is useful when we need to know the total amount of data stored in an array.

# Simple / AI-ML Example

Suppose a dataset contains 5 students and 3 features for each student.

The shape of the dataset would be:

(5, 3)

This means there are 5 rows and 3 columns.

The total number of values is:

5 × 3 = 15

Therefore, the size of the NumPy array is 15.

In AI/ML, knowing the total number of elements helps us understand the amount of data being processed.

```python
import numpy as np

# Create a 2-D array with 5 rows and 3 columns
data = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90],
    [15, 25, 35],
    [45, 55, 65]
])

print("Shape:", data.shape)
print("Total number of elements:", data.size)

# Explanation:
# shape tells us the number of rows and columns.
# size tells us the total number of elements.
# Here, there are 5 rows and 3 columns.
# Therefore, the total number of elements is 5 × 3 = 15.
#
# Expected Output:
# Shape: (5, 3)
# Total number of elements: 15
#
# AI/ML Use:
# The size attribute helps us understand how many individual
# values are present in a dataset or numerical array.
```

# 8. Data Types

A NumPy array stores values using a specific data type called `dtype`.

The data type tells us what kind of values the array contains, such as integers, floating-point numbers, or Boolean values.

NumPy usually assigns a suitable data type automatically when an array is created.

We can check the data type using the `dtype` attribute.

Using an appropriate data type can help reduce memory usage and improve the efficiency of numerical operations.

# Simple / AI-ML Example

Suppose we have the following data:

[10, 20, 30, 40]

# 9. Indexing

Indexing is used to access a specific element from a NumPy array.

NumPy uses zero-based indexing, which means the first element has index `0`, the second element has index `1`, and so on.

For a 1-D array, we use a single index.

For a 2-D array, we can use two indexes: one for the row and one for the column.

Indexing is useful when we need to access or modify specific values in an array.

# Simple / AI-ML Example

Suppose we have the marks of five students:

[80, 75, 90, 85, 95]

If we want to access the third student's mark, we use index `2` because NumPy starts indexing from `0`.

For a dataset with rows and columns, we can use row and column indexes to access a particular value.

In AI/ML, indexing is useful when accessing specific samples, features, or values from numerical data.

```python
import numpy as np

# Create a 1-D NumPy array
marks = np.array([80, 75, 90, 85, 95])

# Access the third element
third_mark = marks[2]

print("Marks:", marks)
print("Third student's mark:", third_mark)

# Create a 2-D NumPy array
student_data = np.array([
    [80, 75, 90],
    [85, 88, 92],
    [70, 78, 82]
])

# Access the value at row 2, column 3
value = student_data[1, 2]

print("\nStudent Data:")
print(student_data)
print("Value at row 2, column 3:", value)

# Explanation:
# NumPy uses zero-based indexing.
# marks[2] accesses the third element.
# student_data[1, 2] accesses the second row and third column.
#
# Expected Output:
# Marks: [80 75 90 85 95]
# Third student's mark: 90
#
# Student Data:
# [[80 75 90]
#  [85 88 92]
#  [70 78 82]]
# Value at row 2, column 3: 92
#
# AI/ML Use:
# Indexing is useful for accessing specific rows, columns,
# samples, or features from numerical datasets.
```

# 10. Slicing

Slicing is used to extract a portion of a NumPy array instead of accessing only one element.

The basic syntax is:

`array[start:stop:step]`

- `start` → index where slicing begins
- `stop` → index where slicing ends (not included)
- `step` → number of positions to move at a time

Slicing can be used with both 1-D and multidimensional arrays.

# Simple / AI-ML Example

Suppose we have daily sales data for one week:

[100, 150, 120, 180, 200, 170, 190]

If we want to analyze only the sales from the first three days, we can use slicing instead of selecting each value separately.

In AI/ML, slicing can be used to select a group of samples, specific columns, or a portion of numerical data for analysis and processing.

```python
import numpy as np

# Create a NumPy array
sales = np.array([100, 150, 120, 180, 200, 170, 190])

# Select the first three values
first_three = sales[0:3]

# Select every second value
every_second = sales[::2]

# Select the last three values
last_three = sales[-3:]

print("Sales:", sales)
print("First three values:", first_three)
print("Every second value:", every_second)
print("Last three values:", last_three)

# Explanation:
# sales[0:3] selects elements from index 0 up to, but not
# including, index 3.
# sales[::2] selects every second element.
# sales[-3:] selects the last three elements.
#
# Expected Output:
# Sales: [100 150 120 180 200 170 190]
# First three values: [100 150 120]
# Every second value: [100 120 200 190]
# Last three values: [200 170 190]
#
# AI/ML Use:
# Slicing is useful for selecting subsets of datasets,
# specific samples, or portions of numerical data for analysis.
```

# 11. Reshaping

Reshaping means changing the structure or dimensions of a NumPy array without changing its elements.

The `reshape()` function is used to change the shape of an array.

For example, an array containing 6 elements can be changed from a 1-D shape `(6,)` into a 2-D shape `(2, 3)`.

The total number of elements must remain the same before and after reshaping.

# Simple / AI-ML Example

Suppose we have six values:

[10, 20, 30, 40, 50, 60]

These values can initially be stored as a 1-D array.

We can reshape them into:

[
    [10, 20, 30],
    [40, 50, 60]
]

The values have not changed; only their structure has changed.

In AI/ML, reshaping is useful when data needs to be converted into the format expected by a Machine Learning algorithm or numerical operation.

```python
import numpy as np

# Create a 1-D array with 6 elements
data = np.array([10, 20, 30, 40, 50, 60])

# Reshape the array into 2 rows and 3 columns
reshaped_data = data.reshape(2, 3)

print("Original Array:", data)
print("Original Shape:", data.shape)

print("\nReshaped Array:")
print(reshaped_data)
print("Reshaped Shape:", reshaped_data.shape)

# Explanation:
# The original array contains 6 elements.
# reshape(2, 3) changes its structure into 2 rows and 3 columns.
# 2 × 3 = 6, so the total number of elements remains unchanged.
#
# Expected Output:
# Original Array: [10 20 30 40 50 60]
# Original Shape: (6,)
#
# Reshaped Array:
# [[10 20 30]
#  [40 50 60]]
# Reshaped Shape: (2, 3)
#
# AI/ML Use:
# Reshaping is commonly used to prepare data in the required
# dimensions before passing it to Machine Learning models.
```

# 12. Flatten

Flattening means converting a multidimensional NumPy array into a one-dimensional array.

The `flatten()` method creates a new 1-D array containing all the elements from the original array.

For example, a 2-D array with rows and columns can be converted into a single sequence of values.

Flattening is useful when data needs to be converted into a single dimension for further processing.

# Simple / AI-ML Example

Suppose we have a 2-D array representing marks:

[
    [80, 90],
    [75, 85]
]

Flattening converts it into:

[80, 90, 75, 85]

The values remain the same; only the structure changes from 2-D to 1-D.

In AI/ML, flattening is commonly used to convert multidimensional data, such as image data, into a one-dimensional format before passing it to certain Machine Learning or Neural Network layers.

```python
import numpy as np

# Create a 2-D NumPy array
data = np.array([
    [80, 90],
    [75, 85]
])

# Convert the 2-D array into a 1-D array
flattened_data = data.flatten()

print("Original Array:")
print(data)

print("\nFlattened Array:")
print(flattened_data)

# Explanation:
# flatten() converts the multidimensional array into a 1-D array.
# It returns a new array containing all the original elements.
#
# Expected Output:
# Original Array:
# [[80 90]
#  [75 85]]
#
# Flattened Array:
# [80 90 75 85]
#
# AI/ML Use:
# Flattening is useful when converting multidimensional data,
# such as image features, into a single-dimensional representation
# for certain Machine Learning operations.
```

# 13. Copy vs View

In NumPy, `copy()` and `view()` are used when creating another array from an existing array, but they behave differently.

A **copy** creates a completely independent array. Changes made to the copy do not affect the original array.

A **view** creates another way of looking at the same underlying data. Changes made through the view can affect the original array.

Understanding the difference is important when working with large datasets because it affects both memory usage and data modification.

# Simple / AI-ML Example

Suppose we have an array containing customer ages:

[20, 25, 30, 35]

If we create a copy and change one value, the original data remains unchanged.

If we create a view and change one value, the original array can also be changed because both arrays share the same underlying data.

In AI/ML, this difference is important when working with large datasets because copies require additional memory, while views can help avoid unnecessary data duplication.

```python
import numpy as np

# Original array
original = np.array([10, 20, 30, 40])

# Create a copy
copied_array = original.copy()

# Change the copy
copied_array[0] = 100

print("Original after changing copy:", original)
print("Copy:", copied_array)

# Create a view
view_array = original.view()

# Change the view
view_array[1] = 200

print("\nOriginal after changing view:", original)
print("View:", view_array)

# Explanation:
# copy() creates an independent array.
# Therefore, changing copied_array does not change original.
#
# view() creates another view of the same underlying data.
# Therefore, changing view_array also changes original.
#
# Expected Output:
# Original after changing copy: [10 20 30 40]
# Copy: [100  20  30  40]
#
# Original after changing view: [ 10 200  30  40]
# View: [ 10 200  30  40]
#
# AI/ML Use:
# Understanding copy and view helps avoid accidental changes
# to datasets and can also help manage memory efficiently
# when working with large numerical datasets.
```

# 14. Broadcasting

Broadcasting is a NumPy feature that allows arithmetic operations between arrays with different shapes when their shapes are compatible.

Instead of manually repeating or reshaping values, NumPy automatically adjusts the smaller array during the operation.

For example, we can add a single value to every element of an array.

Broadcasting makes numerical calculations simpler and avoids unnecessary loops.

# Simple / AI-ML Example

Suppose we have the prices of three products:

[100, 200, 300]

If we want to add a tax of 10 to every product, we can simply add 10 to the entire NumPy array.

NumPy broadcasts the single value `10` to every element:

[100, 200, 300] + 10

becomes:

[110, 210, 310]

In AI/ML, broadcasting is useful when applying the same value, adjustment, or transformation across many data points or features.

```python
import numpy as np

# Create an array containing product prices
prices = np.array([100, 200, 300])

# Add the same tax amount to every price
tax = 10
final_prices = prices + tax

print("Original Prices:", prices)
print("Tax:", tax)
print("Final Prices:", final_prices)

# Explanation:
# prices is a NumPy array containing three values.
# tax is a single value.
# NumPy broadcasts the value 10 across all elements of prices.
#
# Expected Output:
# Original Prices: [100 200 300]
# Tax: 10
# Final Prices: [110 210 310]
#
# AI/ML Use:
# Broadcasting is commonly used to apply the same mathematical
# operation to many values, such as adding offsets, scaling data,
# or performing feature transformations.
```

# 15. Array Operations

NumPy allows us to perform arithmetic operations directly on arrays.

Common array operations include:

- Addition
- Subtraction
- Multiplication
- Division
- Power

These operations are usually performed element by element when the arrays have compatible shapes.

NumPy makes these operations simple and efficient without requiring explicit loops.

# Simple / AI-ML Example

Suppose we have the sales of two products:

Product A: [100, 200, 300]
Product B: [50, 100, 150]

Using NumPy, we can add the arrays to calculate the combined sales:

[100, 200, 300] + [50, 100, 150]

Result:

[150, 300, 450]

Similarly, we can subtract, multiply, or divide corresponding values.

In AI/ML, array operations are frequently used for calculations involving features, predictions, transformations, and numerical datasets.

```python
import numpy as np

# Create two NumPy arrays
product_a = np.array([100, 200, 300])
product_b = np.array([50, 100, 150])

# Perform arithmetic operations
addition = product_a + product_b
subtraction = product_a - product_b
multiplication = product_a * product_b
division = product_a / product_b

print("Addition:", addition)
print("Subtraction:", subtraction)
print("Multiplication:", multiplication)
print("Division:", division)

# Explanation:
# Addition adds corresponding elements.
# Subtraction subtracts corresponding elements.
# Multiplication multiplies corresponding elements.
# Division divides corresponding elements.
#
# Expected Output:
# Addition: [150 300 450]
# Subtraction: [ 50 100 150]
# Multiplication: [ 5000 20000 45000]
# Division: [2. 2. 2.]
#
# AI/ML Use:
# Array operations are used extensively for mathematical
# calculations, feature transformations, predictions, and
# numerical processing in Machine Learning.
```

# 16. Universal Functions (ufuncs)

Universal functions, commonly called `ufuncs`, are NumPy functions that perform operations element by element on NumPy arrays.

They can work with individual values as well as complete arrays and are designed to perform numerical operations efficiently.

Examples of NumPy ufuncs include:

- `np.sqrt()` – calculates square root
- `np.exp()` – calculates exponential
- `np.sin()` – calculates sine
- `np.log()` – calculates natural logarithm
- `np.abs()` – calculates absolute value

Ufuncs help us perform mathematical operations on entire arrays without writing explicit loops.

# Simple / AI-ML Example

Suppose we have the following values:

[1, 4, 9, 16]

If we want to calculate the square root of every value, we can use `np.sqrt()` directly on the array.

The result will be:

[1, 2, 3, 4]

Instead of calculating each value separately using a loop, the NumPy ufunc performs the operation on the complete array.

In AI/ML, ufuncs are useful for mathematical transformations and numerical preprocessing of data.

```python
import numpy as np

# Create a NumPy array
numbers = np.array([1, 4, 9, 16])

# Apply the square root universal function
square_roots = np.sqrt(numbers)

# Apply the absolute value universal function
values = np.array([-10, -5, 5, 10])
absolute_values = np.abs(values)

print("Numbers:", numbers)
print("Square Roots:", square_roots)

print("\nValues:", values)
print("Absolute Values:", absolute_values)

# Explanation:
# np.sqrt() is a NumPy universal function that calculates
# the square root of each element.
# np.abs() calculates the absolute value of each element.
# Both functions operate element by element on the array.
#
# Expected Output:
# Numbers: [ 1  4  9 16]
# Square Roots: [1. 2. 3. 4.]
#
# Values: [-10  -5   5  10]
# Absolute Values: [10  5  5 10]
#
# AI/ML Use:
# Ufuncs are useful for mathematical transformations,
# feature preprocessing, and numerical calculations
# performed during Data Analysis and Machine Learning.
```
