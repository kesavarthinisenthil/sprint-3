# 2. Mean

Mean is the average value of a set of numbers.

In NumPy, we use `np.mean()` to calculate the mean of an array.

The formula is:

Mean = Sum of all values / Number of values

Mean helps us understand the central or average value of numerical data.

In AI/ML, mean is commonly used to understand numerical features and during data preprocessing.

# Simple / AI-ML Example

Suppose the marks of five students are:

[60, 70, 80, 90, 100]

The mean gives us the average mark.

Mean = (60 + 70 + 80 + 90 + 100) / 5
     = 80

So, the average mark is 80.

In AI/ML, mean can be used to understand the average value of a feature, such as average age, average salary, or average sales.

```python
import numpy as np

# Create a NumPy array containing student marks
marks = np.array([60, 70, 80, 90, 100])

# Calculate the mean using np.mean()
average_marks = np.mean(marks)

print("Marks:", marks)
print("Mean:", average_marks)

# Explanation:
# np.mean() calculates the average of all values in the array.
# It adds all the values and divides them by the number of values.
#
# Expected Output:
# Marks: [ 60  70  80  90 100]
# Mean: 80.0
#
# AI/ML Use:
# Mean is useful for understanding the average value of
# numerical features during data analysis and preprocessing.
```

# 3. Median

Median is the middle value of a set of numbers after arranging the values in ascending order.

In NumPy, we use `np.median()` to calculate the median.

If there is an odd number of values, the middle value is the median.

If there is an even number of values, the median is the average of the two middle values.

Median is useful when the data contains extreme values or outliers because it is less affected by them than the mean.

# Simple / AI-ML Example

Suppose the salaries of five employees are:

[20000, 22000, 25000, 28000, 100000]

After arranging the values, the middle value is:

25000

So, the median salary is 25000.

Here, 100000 is an unusually high value. It affects the mean strongly, but the median still represents the typical salary better.

In AI/ML and Data Analytics, median is useful for understanding numerical data when outliers are present.

```python
import numpy as np

# Create a NumPy array containing employee salaries
salaries = np.array([20000, 22000, 25000, 28000, 100000])

# Calculate the median using np.median()
median_salary = np.median(salaries)

print("Salaries:", salaries)
print("Median Salary:", median_salary)

# Explanation:
# np.median() finds the middle value of the array
# after arranging the values in ascending order.
#
# Expected Output:
# Salaries: [ 20000  22000  25000  28000 100000]
# Median Salary: 25000.0
#
# AI/ML Use:
# Median is useful when a dataset contains outliers.
# It can be used to understand typical values during
# data analysis and data preprocessing.
```

# 4. Variance

Variance measures how much the values in a dataset differ from the mean.

A small variance means the values are close to the mean.

A large variance means the values are more spread out from the mean.

In NumPy, we use `np.var()` to calculate variance.

Variance is useful for understanding the spread of numerical data and how much a feature varies.

# Simple / AI-ML Example

Suppose we have two sets of student marks:

Dataset A:
[48, 49, 50, 51, 52]

Dataset B:
[20, 35, 50, 65, 80]

Both datasets have a similar average, but Dataset B has values that are much more spread out.

Therefore, Dataset B will have a higher variance.

In AI/ML, variance helps us understand how much a feature changes across the dataset and is useful during data analysis and feature preprocessing.

```python
import numpy as np

# Create a NumPy array containing student marks
marks = np.array([48, 49, 50, 51, 52])

# Calculate the variance using np.var()
variance = np.var(marks)

print("Marks:", marks)
print("Variance:", variance)

# Explanation:
# np.var() calculates how much the values differ
# from the mean of the dataset.
#
# Expected Output:
# Marks: [48 49 50 51 52]
# Variance: 2.0
#
# AI/ML Use:
# Variance helps us understand the spread of a feature.
# It can also help identify features with very low or
# very high variation during data preprocessing.
```

# 5. Standard Deviation

Standard deviation measures how much the values in a dataset are spread out from the mean.

A small standard deviation means the values are close to the mean.

A large standard deviation means the values are more spread out from the mean.

In NumPy, we use `np.std()` to calculate standard deviation.

Standard deviation is the square root of variance.

In AI/ML, standard deviation is commonly used to understand data spread and for feature scaling and normalization.

# Simple / AI-ML Example

Suppose the marks of five students are:

[48, 49, 50, 51, 52]

The values are very close to the average, so the standard deviation will be small.

If the marks were:

[20, 35, 50, 65, 80]

The values would be more spread out, so the standard deviation would be larger.

In AI/ML, standard deviation helps us understand how much a numerical feature varies and is useful during feature scaling.

```python
import numpy as np

# Create a NumPy array containing student marks
marks = np.array([48, 49, 50, 51, 52])

# Calculate the standard deviation using np.std()
standard_deviation = np.std(marks)

print("Marks:", marks)
print("Standard Deviation:", standard_deviation)

# Explanation:
# np.std() calculates the standard deviation of the values.
# It tells us how much the values are spread around the mean.
#
# Expected Output:
# Marks: [48 49 50 51 52]
# Standard Deviation: 1.4142135623730951
#
# AI/ML Use:
# Standard deviation is useful for understanding feature
# variation and is commonly used in feature scaling,
# normalization, and data preprocessing.
```

# 8. Argmin

`argmin` returns the **index (position)** of the smallest value in an array.

In NumPy, we use `np.argmin()` to find the index of the minimum value.

Remember:

- `np.min()` → returns the smallest value
- `np.argmin()` → returns the position of the smallest value

The index starts from `0` in Python.

In AI/ML and Data Analytics, `argmin` can help identify the position of the feature, observation, or value with the lowest measurement.

# Simple / AI-ML Example

Suppose the prices of four products are:

[500, 300, 450, 250]

The minimum price is:

250

Its index is:

3

So:

`np.min()` → 250

`np.argmin()` → 3

In AI/ML, `argmin` can be useful when we want to find which position contains the lowest value, such as the lowest prediction error or minimum cost.

```python
import numpy as np

# Create a NumPy array containing product prices
prices = np.array([500, 300, 450, 250])

# Find the index of the minimum value using np.argmin()
minimum_index = np.argmin(prices)

# Find the minimum value using np.min()
minimum_price = np.min(prices)

print("Prices:", prices)
print("Minimum Price:", minimum_price)
print("Index of Minimum Price:", minimum_index)

# Explanation:
# np.min() returns the smallest value in the array.
# np.argmin() returns the index of the smallest value.
#
# Expected Output:
# Prices: [500 300 450 250]
# Minimum Price: 250
# Index of Minimum Price: 3
#
# AI/ML Use:
# argmin can be used to identify the position of the
# lowest error, cost, distance, or other numerical value.
```

# 9. Argmax

`argmax` returns the **index (position)** of the largest value in an array.

In NumPy, we use `np.argmax()` to find the index of the maximum value.

Remember:

- `np.max()` → returns the largest value
- `np.argmax()` → returns the position of the largest value

The index starts from `0` in Python.

In AI/ML and Data Analytics, `argmax` is useful when we want to identify the position of the highest value, such as the class with the highest prediction probability.

# Simple / AI-ML Example

Suppose an AI model gives prediction probabilities for three classes:

[0.10, 0.75, 0.15]

The highest probability is:

0.75

Its index is:

1

So:

`np.max()` → 0.75

`np.argmax()` → 1

In AI/ML, `argmax` is commonly used to select the class with the highest prediction probability.

```python
import numpy as np

# Prediction probabilities for three classes
probabilities = np.array([0.10, 0.75, 0.15])

# Find the index of the highest probability
maximum_index = np.argmax(probabilities)

# Find the highest probability
maximum_probability = np.max(probabilities)

print("Prediction Probabilities:", probabilities)
print("Maximum Probability:", maximum_probability)
print("Index of Maximum Probability:", maximum_index)

# Explanation:
# np.max() returns the highest value in the array.
# np.argmax() returns the index of the highest value.
#
# Expected Output:
# Prediction Probabilities: [0.1  0.75 0.15]
# Maximum Probability: 0.75
# Index of Maximum Probability: 1
#
# AI/ML Use:
# argmax is commonly used in classification models
# to select the class having the highest prediction probability.
```

# 10. Percentile

A percentile tells us the value below which a certain percentage of the data falls.

In NumPy, we use `np.percentile()` to calculate a percentile.

For example:

- 25th percentile → 25% of the values are at or below this value.
- 50th percentile → 50% of the values are at or below this value. This is the median.
- 75th percentile → 75% of the values are at or below this value.

Percentiles are useful for understanding the distribution and position of values in a dataset.

In AI/ML and Data Analytics, percentiles are commonly used for understanding data distribution and detecting outliers.

# Simple / AI-ML Example

Suppose we have the following exam scores:

[40, 50, 60, 70, 80, 90, 100]

If we calculate the 50th percentile, it represents the middle of the data.

The 50th percentile is:

70

So, approximately 50% of the scores are at or below 70.

In AI/ML and Data Analytics, percentiles can be used to understand feature distributions and identify unusually high or low values.

```python
import numpy as np

# Create a NumPy array containing exam scores
scores = np.array([40, 50, 60, 70, 80, 90, 100])

# Calculate the 50th percentile
percentile_50 = np.percentile(scores, 50)

# Calculate the 75th percentile
percentile_75 = np.percentile(scores, 75)

print("Scores:", scores)
print("50th Percentile:", percentile_50)
print("75th Percentile:", percentile_75)

# Explanation:
# np.percentile() calculates the value below which
# a specified percentage of the data falls.
#
# Expected Output:
# Scores: [ 40  50  60  70  80  90 100]
# 50th Percentile: 70.0
# 75th Percentile: 85.0
#
# AI/ML Use:
# Percentiles are useful for understanding data distribution,
# identifying extreme values, and detecting potential outliers.
```

# 11. Dot Product

The dot product is a mathematical operation that multiplies corresponding elements of two arrays and then adds all the results.

In NumPy, we can use `np.dot()` to calculate the dot product.

For example:

Array A = [1, 2, 3]
Array B = [4, 5, 6]

Dot Product = (1 × 4) + (2 × 5) + (3 × 6)
            = 4 + 10 + 18
            = 32

In AI/ML, dot products are widely used in machine learning calculations, especially in linear models, neural networks, and feature-weight calculations.

# Simple / AI-ML Example

Suppose we have three features:

Features = [2, 3, 4]

And their corresponding weights are:

Weights = [0.5, 1.0, 1.5]

The dot product combines the features and their weights:

(2 × 0.5) + (3 × 1.0) + (4 × 1.5)
= 1 + 3 + 6
= 10

So, the weighted result is 10.

In AI/ML, this type of calculation is commonly used when a model combines input features with their weights to produce a prediction or score.

```python
import numpy as np

# Create an array containing feature values
features = np.array([2, 3, 4])

# Create an array containing corresponding weights
weights = np.array([0.5, 1.0, 1.5])

# Calculate the dot product of features and weights
result = np.dot(features, weights)

print("Features:", features)
print("Weights:", weights)
print("Dot Product:", result)

# Explanation:
# np.dot() multiplies corresponding elements of two arrays
# and adds the results together.
#
# Expected Output:
# Features: [2 3 4]
# Weights: [0.5 1.  1.5]
# Dot Product: 10.0
#
# AI/ML Use:
# Dot products are fundamental in machine learning.
# They are used to combine input features with model weights
# in linear regression, classification, and neural networks.
```

# 12. Matrix Multiplication

Matrix multiplication is an operation where rows of the first matrix are multiplied with columns of the second matrix and the results are added together.

In NumPy, we can use `np.matmul()` or the `@` operator for matrix multiplication.

For matrix multiplication:

- The number of columns in the first matrix must equal the number of rows in the second matrix.
- The resulting matrix contains the calculated row-column products.

Matrix multiplication is very important in AI/ML because many machine learning and deep learning operations use matrices.

# Simple / AI-ML Example

Suppose we have:

Matrix A:

[[1, 2],
 [3, 4]]

Matrix B:

[[5, 6],
 [7, 8]]

The first value of the result is calculated using the first row of A and first column of B:

(1 × 5) + (2 × 7) = 19

The complete result is:

[[19, 22],
 [43, 50]]

In AI/ML, matrix multiplication is used to combine input data with weights in machine learning models and neural networks.

```python
import numpy as np

# Create the first matrix
A = np.array([
    [1, 2],
    [3, 4]
])

# Create the second matrix
B = np.array([
    [5, 6],
    [7, 8]
])

# Perform matrix multiplication using np.matmul()
result = np.matmul(A, B)

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)

print("\nMatrix Multiplication Result:")
print(result)

# Explanation:
# np.matmul() performs matrix multiplication.
# Each value in the result is calculated by multiplying
# a row of A with a column of B and adding the results.
#
# Expected Output:
# Matrix A:
# [[1 2]
#  [3 4]]
#
# Matrix B:
# [[5 6]
#  [7 8]]
#
# Matrix Multiplication Result:
# [[19 22]
#  [43 50]]
#
# AI/ML Use:
# Matrix multiplication is widely used in machine learning
# algorithms and neural networks to perform calculations
# between input data and model weights.
```

# 13. Transpose

Transpose changes the rows of a matrix into columns and the columns into rows.

In NumPy, we can use `.T` or `np.transpose()` to find the transpose of an array.

For example:

Original matrix:

[[1, 2, 3],
 [4, 5, 6]]

Transpose:

[[1, 4],
 [2, 5],
 [3, 6]]

Transpose is useful when we need to change the orientation of data or perform matrix operations.

In AI/ML, transpose is commonly used during matrix calculations, feature transformations, and linear algebra operations.

# Simple / AI-ML Example

Suppose a dataset contains:

Rows → Students  
Columns → Subjects

For example:

[[80, 75, 90],
 [70, 85, 88]]

Here, each row represents a student and each column represents a subject.

After transpose:

[[80, 70],
 [75, 85],
 [90, 88]]

Now, the rows represent subjects and the columns represent students.

In AI/ML, transposing data can be useful when changing the orientation of matrices for mathematical calculations.

```python
import numpy as np

# Create a 2D NumPy array
marks = np.array([
    [80, 75, 90],
    [70, 85, 88]
])

# Find the transpose using .T
transposed_marks = marks.T

print("Original Matrix:")
print(marks)

print("\nTransposed Matrix:")
print(transposed_marks)

# Explanation:
# .T changes rows into columns and columns into rows.
#
# Expected Output:
# Original Matrix:
# [[80 75 90]
#  [70 85 88]]
#
# Transposed Matrix:
# [[80 70]
#  [75 85]
#  [90 88]]
#
# AI/ML Use:
# Transpose is useful in matrix calculations,
# feature transformations, and preparing data
# for mathematical operations in machine learning.
```

# 14. Basic Linear Algebra Operations

Linear algebra is the study of vectors, matrices, and mathematical operations between them.

NumPy provides several functions for basic linear algebra operations.

Some commonly used operations are:

- `np.dot()` – calculates the dot product
- `np.matmul()` – performs matrix multiplication
- `np.transpose()` – changes rows into columns
- `np.linalg.det()` – calculates the determinant of a matrix
- `np.linalg.inv()` – calculates the inverse of a matrix

These operations are important for working with matrices and vectors.

In AI/ML, linear algebra is widely used in machine learning algorithms, neural networks, transformations, optimization, and handling numerical data.

# Simple / AI-ML Example

Suppose we have a matrix:

[[2, 1],
 [1, 2]]

We can perform different linear algebra operations on this matrix.

For example:

- Calculate its determinant.
- Calculate its inverse.
- Transpose the matrix.

These operations help us understand and manipulate matrix data.

In AI/ML, linear algebra operations are used internally in many algorithms to perform calculations on features, weights, and model parameters.

```python
import numpy as np

# Create a 2x2 matrix
matrix = np.array([
    [2, 1],
    [1, 2]
])

# Calculate the determinant
determinant = np.linalg.det(matrix)

# Calculate the inverse
inverse = np.linalg.inv(matrix)

# Calculate the transpose
transpose = np.transpose(matrix)

print("Matrix:")
print(matrix)

print("\nDeterminant:")
print(determinant)

print("\nInverse:")
print(inverse)

print("\nTranspose:")
print(transpose)

# Explanation:
# np.linalg.det() calculates the determinant of a matrix.
# np.linalg.inv() calculates the inverse of a matrix.
# np.transpose() changes rows into columns.
#
# Expected Output:
# Matrix:
# [[2 1]
#  [1 2]]
#
# Determinant:
# 3.0
#
# Inverse:
# [[ 0.66666667 -0.33333333]
#  [-0.33333333  0.66666667]]
#
# Transpose:
# [[2 1]
#  [1 2]]
#
# AI/ML Use:
# Linear algebra operations are fundamental to AI/ML.
# They are used for matrix calculations, transformations,
# optimization, and calculations inside machine learning
# and neural network algorithms.
```
