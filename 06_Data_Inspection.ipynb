# head()

## Understand the Concept

The `head()` function is used to view the first few rows of a Pandas DataFrame.

By default, `head()` displays the first 5 rows.

We can also specify the number of rows we want to see, such as `head(3)` to display the first 3 rows.

It is mainly used to quickly understand what the dataset looks like before performing data analysis or preprocessing.

## Demonstrate the Concept

### Real-World Example

Suppose we have a customer dataset containing thousands of records.

Instead of displaying all the rows, we can use `head()` to quickly check the first few records and understand the structure of the data.

### AI/ML Example

Before building a Machine Learning model, we can use `head()` to quickly inspect the dataset and check whether the columns and values look correct.

For example, we can check whether columns such as `Age`, `Salary`, and `Purchased` contain the expected data.

```python
import pandas as pd

# Create a sample customer DataFrame.
data = {
    "Name": ["Arun", "Priya", "Kumar", "Divya", "Ravi", "Anu"],
    "Age": [25, 28, 30, 24, 32, 27],
    "Purchase": [500, 750, 1000, 450, 1200, 800]
}

df = pd.DataFrame(data)

# Display the first 5 rows.
# head() is used to quickly inspect the beginning of the dataset.
print(df.head())

# Display the first 3 rows.
# We pass 3 as an argument when we need a specific number of rows.
print("\nFirst 3 rows:")
print(df.head(3))

# Expected Output for head():
#     Name  Age  Purchase
# 0   Arun   25       500
# 1  Priya   28       750
# 2  Kumar   30      1000
# 3  Divya   24       450
# 4   Ravi   32      1200

# AI/ML / Data Analytics Use:
# head() is commonly used during Exploratory Data Analysis (EDA)
# to quickly understand the structure and initial values of a dataset.
```

# tail()

## Understand the Concept

The `tail()` function is used to view the last few rows of a Pandas DataFrame.

By default, `tail()` displays the last 5 rows.

We can also specify the number of rows we want to see, such as `tail(3)` to display the last 3 rows.

It is useful for checking the ending records of a dataset and verifying whether the data was loaded or processed correctly.

## Demonstrate the Concept

### Real-World Example

Suppose a company has a large sales dataset.

We can use `tail()` to check the last few sales records and make sure the dataset has been loaded completely.

### AI/ML Example

During data preprocessing, we can use `tail()` to check the last records after sorting, filtering, or adding new data.

This helps us verify that the final rows contain the expected values.

```python
import pandas as pd

# Create a sample customer DataFrame.
data = {
    "Name": ["Arun", "Priya", "Kumar", "Divya", "Ravi", "Anu"],
    "Age": [25, 28, 30, 24, 32, 27],
    "Purchase": [500, 750, 1000, 450, 1200, 800]
}

df = pd.DataFrame(data)

# Display the last 5 rows.
# tail() is used to quickly inspect the end of the dataset.
print(df.tail())

# Display the last 3 rows.
# We pass 3 when we want to view a specific number of rows.
print("\nLast 3 rows:")
print(df.tail(3))

# Expected Output for tail():
#     Name  Age  Purchase
# 1  Priya   28       750
# 2  Kumar   30      1000
# 3  Divya   24       450
# 4   Ravi   32      1200
# 5    Anu   27       800

# AI/ML / Data Analytics Use:
# tail() is useful during data inspection and EDA
# to verify the last records of a dataset.
```

# sample()

## Understand the Concept

The `sample()` function is used to randomly select rows from a Pandas DataFrame.

By default, `sample()` returns one random row.

We can specify the number of rows we want to select, such as `sample(3)` to randomly select 3 rows.

It is useful when we want to inspect random records instead of only looking at the beginning or end of the dataset.

## Demonstrate the Concept

### Real-World Example

Suppose a company has thousands of customer records.

Instead of checking every record, we can randomly select a few records using `sample()` to get a quick idea of the data.

### AI/ML Example

In Machine Learning and Data Analytics, `sample()` can be useful for randomly inspecting a dataset or creating a smaller sample for testing and exploration.

Random sampling can help us check whether the dataset contains unexpected values or patterns in different parts of the data.

```python
import pandas as pd

# Create a sample customer DataFrame.
data = {
    "Name": ["Arun", "Priya", "Kumar", "Divya", "Ravi", "Anu"],
    "Age": [25, 28, 30, 24, 32, 27],
    "Purchase": [500, 750, 1000, 450, 1200, 800]
}

df = pd.DataFrame(data)

# Select one random row.
# sample() randomly selects rows from the DataFrame.
print("One random row:")
print(df.sample())

# Select three random rows.
# sample(3) randomly selects 3 rows.
print("\nThree random rows:")
print(df.sample(3))

# Expected Output:
# The output will contain randomly selected rows.
# The rows may be different each time the code is executed.

# AI/ML / Data Analytics Use:
# sample() is useful for random data inspection,
# exploratory data analysis, and creating smaller samples
# from large datasets.
```

# shape

## Understand the Concept

The `shape` attribute is used to find the number of rows and columns in a Pandas DataFrame.

It returns a tuple in the format:

`(number_of_rows, number_of_columns)`

For example, `(100, 5)` means the DataFrame contains 100 rows and 5 columns.

`shape` is an attribute, so we use `df.shape` without parentheses.

## Demonstrate the Concept

### Real-World Example

Suppose a company has a customer dataset with 10,000 customers and 8 columns.

Using `shape`, we can quickly check whether the dataset contains 10,000 rows and 8 columns.

### AI/ML Example

In Machine Learning, `shape` helps us understand the size of the dataset.

For example, before training a model, we can check how many samples and features are available.

A dataset with `(1000, 10)` contains 1,000 rows and 10 columns.

```python
import pandas as pd

# Create a sample DataFrame.
data = {
    "Name": ["Arun", "Priya", "Kumar", "Divya", "Ravi", "Anu"],
    "Age": [25, 28, 30, 24, 32, 27],
    "Purchase": [500, 750, 1000, 450, 1200, 800]
}

df = pd.DataFrame(data)

# Get the number of rows and columns.
# shape returns (rows, columns).
print("DataFrame Shape:", df.shape)

# Get the number of rows.
print("Number of Rows:", df.shape[0])

# Get the number of columns.
print("Number of Columns:", df.shape[1])

# Expected Output:
# DataFrame Shape: (6, 3)
# Number of Rows: 6
# Number of Columns: 3

# AI/ML / Data Analytics Use:
# shape is useful for checking dataset size and
# understanding the number of samples and features
# before data preprocessing and model training.
```

# columns

## Understand the Concept

The `columns` attribute is used to get the names of all columns in a Pandas DataFrame.

It returns the column names as a Pandas Index.

We can use it to:
- Check what features are available
- Access column names
- Verify whether expected columns exist
- Understand the structure of a dataset

`columns` is an attribute, so we use `df.columns` without parentheses.

## Demonstrate the Concept

### Real-World Example

Suppose a customer dataset contains columns such as `Name`, `Age`, `Purchase`, and `City`.

Using `columns`, we can quickly check all the available fields in the dataset.

### AI/ML Example

In Machine Learning, the columns often represent features or target variables.

For example, `Age`, `Income`, and `Purchase_History` can be features used to predict whether a customer will buy a product.

Using `columns` helps us identify and select the required features.

```python
import pandas as pd

# Create a sample DataFrame.
data = {
    "Name": ["Arun", "Priya", "Kumar"],
    "Age": [25, 28, 30],
    "Purchase": [500, 750, 1000]
}

df = pd.DataFrame(data)

# Display all column names.
# columns returns the names of all columns in the DataFrame.
print("Column Names:")
print(df.columns)

# Convert column names into a Python list.
# tolist() makes the column names easier to work with.
print("\nColumn Names as List:")
print(df.columns.tolist())

# Expected Output:
# Column Names:
# Index(['Name', 'Age', 'Purchase'], dtype='object')
#
# Column Names as List:
# ['Name', 'Age', 'Purchase']

# AI/ML / Data Analytics Use:
# columns helps us identify available features and select
# the required columns for data analysis and Machine Learning.
```

# dtypes

## Understand the Concept

The `dtypes` attribute is used to check the data type of each column in a Pandas DataFrame.

Common data types include:
- `int64` – integer numbers
- `float64` – decimal numbers
- `object` or `string` – text values
- `bool` – True or False values
- `datetime` – date and time values

Understanding data types is important because different types of data require different operations and preprocessing techniques.

## Demonstrate the Concept

### Real-World Example

Suppose a customer dataset contains:
- `Name` – text
- `Age` – integer
- `Purchase` – decimal value
- `Subscribed` – True/False

Using `dtypes`, we can check whether each column has the correct data type.

### AI/ML Example

Before training a Machine Learning model, we need to understand the data types of the features.

For example, numerical features can be used for mathematical calculations, while categorical or text features may need encoding before being given to a model.

```python
import pandas as pd

# Create a sample DataFrame.
data = {
    "Name": ["Arun", "Priya", "Kumar"],
    "Age": [25, 28, 30],
    "Purchase": [500.50, 750.75, 1000.25],
    "Subscribed": [True, False, True]
}

df = pd.DataFrame(data)

# Display the data type of each column.
# dtypes helps us understand how Pandas stores each column.
print("Data Types:")
print(df.dtypes)

# Expected Output:
# Name            object
# Age              int64
# Purchase       float64
# Subscribed        bool
#
# The exact integer/float dtype may vary depending on
# the Pandas and Python environment.

# AI/ML / Data Analytics Use:
# dtypes helps identify numerical, categorical, text,
# and Boolean features before data preprocessing and
# Machine Learning model training.
```

# memory_usage()

## Understand the Concept

The `memory_usage()` method is used to check how much memory each column of a Pandas DataFrame is using.

It returns the memory usage in bytes.

We can use `memory_usage()` to:
- Understand memory consumption
- Identify columns that use more memory
- Optimize large datasets
- Improve the efficiency of data processing

Using `deep=True` gives a more accurate memory usage for object or string columns.

## Demonstrate the Concept

### Real-World Example

Suppose a company has a dataset containing millions of customer records.

Some columns may consume a large amount of memory, especially text columns.

Using `memory_usage()`, we can identify memory-consuming columns and optimize the dataset.

### AI/ML Example

AI/ML projects often work with large datasets.

Reducing unnecessary memory usage can make data processing faster and can help prevent memory-related problems when training or preprocessing data.

For example, converting suitable columns to more efficient data types can reduce memory consumption.

```python
import pandas as pd

# Create a sample DataFrame.
data = {
    "Name": ["Arun", "Priya", "Kumar", "Divya"],
    "Age": [25, 28, 30, 24],
    "Purchase": [500.50, 750.75, 1000.25, 450.00]
}

df = pd.DataFrame(data)

# Check memory usage of each column.
# memory_usage() returns the memory used by each column in bytes.
print("Memory Usage:")
print(df.memory_usage())

# Check memory usage including the actual memory used
# by object/string values.
print("\nDetailed Memory Usage:")
print(df.memory_usage(deep=True))

# Expected Output:
# The output shows the memory used by the DataFrame index
# and each individual column in bytes.
#
# The exact values may vary depending on the data and environment.

# AI/ML / Data Analytics Use:
# memory_usage() is useful when working with large datasets.
# It helps identify memory-heavy columns so we can optimize
# data types and improve processing efficiency.
```

# nunique()

## Understand the Concept

The `nunique()` method is used to find the number of unique values in each column of a Pandas DataFrame.

For example, if a column contains:

`["Chennai", "Coimbatore", "Chennai", "Salem"]`

there are 3 unique values:

`Chennai`, `Coimbatore`, and `Salem`.

By default, `nunique()` does not count missing values.

## Demonstrate the Concept

### Real-World Example

Suppose a company has customer data with a `City` column.

Using `nunique()`, we can find how many different cities are present in the dataset.

This helps us understand the variety of values in categorical columns.

### AI/ML Example

In Machine Learning, `nunique()` is useful for understanding categorical features.

For example, if a `Department` column has only 4 unique values, it may be treated as a categorical feature and can later be encoded for a Machine Learning model.

```python
import pandas as pd

# Create a sample DataFrame.
data = {
    "Name": ["Arun", "Priya", "Kumar", "Divya", "Ravi"],
    "City": ["Chennai", "Coimbatore", "Chennai", "Salem", "Coimbatore"],
    "Age": [25, 28, 30, 24, 32]
}

df = pd.DataFrame(data)

# Find the number of unique values in each column.
# nunique() counts the different values present in each column.
print("Number of Unique Values:")
print(df.nunique())

# Find the number of unique cities.
print("\nUnique Cities:")
print(df["City"].nunique())

# Expected Output:
# Number of Unique Values:
# Name    5
# City    3
# Age     5
#
# Unique Cities:
# 3

# AI/ML / Data Analytics Use:
# nunique() helps identify categorical columns and understand
# the number of distinct categories before data preprocessing
# and feature encoding.
```

# value_counts()

## Understand the Concept

The `value_counts()` method is used to count how many times each unique value appears in a Pandas Series or column.

It is mainly used to understand the frequency or distribution of categorical values.

For example, if a `City` column contains:

`Chennai, Chennai, Salem, Chennai, Salem`

`value_counts()` will show:

- Chennai → 3
- Salem → 2

By default, the results are sorted from the most frequent value to the least frequent value.

## Demonstrate the Concept

### Real-World Example

Suppose a company has customer information containing a `City` column.

Using `value_counts()`, we can find how many customers belong to each city.

This helps the company understand where most of its customers are located.

### AI/ML Example

In Machine Learning, `value_counts()` is useful for understanding the distribution of categorical variables and target classes.

For example, if a classification target contains `Yes` and `No`, we can use `value_counts()` to check how many records belong to each class.

This can help identify class imbalance before training a model.

```python
import pandas as pd

# Create a sample customer DataFrame.
data = {
    "Name": ["Arun", "Priya", "Kumar", "Divya", "Ravi", "Anu"],
    "City": ["Chennai", "Coimbatore", "Chennai", "Salem", "Coimbatore", "Chennai"],
    "Purchased": ["Yes", "No", "Yes", "No", "Yes", "Yes"]
}

df = pd.DataFrame(data)

# Count how many times each city appears.
# value_counts() counts the frequency of each unique value.
print("City Frequency:")
print(df["City"].value_counts())

# Count the values in the Purchased column.
print("\nPurchase Frequency:")
print(df["Purchased"].value_counts())

# Expected Output:
# City Frequency:
# Chennai       3
# Coimbatore    2
# Salem         1
#
# Purchase Frequency:
# Yes    4
# No     2

# AI/ML / Data Analytics Use:
# value_counts() is useful for understanding categorical
# distributions and checking class balance in classification
# datasets before Machine Learning model training.
```
