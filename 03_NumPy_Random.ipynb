# 1. Random Numbers

Random numbers are numbers generated in an unpredictable or random-like manner.

NumPy provides the `np.random` module to generate random numbers.

We can use `np.random.rand()` to generate random floating-point numbers between 0 and 1.

Random numbers are useful for testing, simulations, sampling, and creating example datasets.

In AI/ML, random numbers are commonly used for initializing model parameters, creating random samples, and splitting or shuffling data.

# Simple / AI-ML Example

Suppose we want to create random values representing the probability of an event.

For example:

[0.23, 0.71, 0.45, 0.89]

Each value is randomly generated between 0 and 1.

Instead of manually creating these values, NumPy can generate them automatically.

In AI/ML, random numbers are useful when creating sample data, initializing model parameters, and performing simulations.

```python
import numpy as np

# Generate 5 random floating-point numbers between 0 and 1
random_numbers = np.random.rand(5)

print("Random Numbers:", random_numbers)

# Explanation:
# np.random.rand(5) generates 5 random floating-point
# numbers between 0 and 1.
#
# The exact output will be different each time
# because the numbers are randomly generated.
#
# Example Output:
# Random Numbers: [0.42 0.17 0.83 0.65 0.29]
#
# AI/ML Use:
# Random numbers are used for creating sample data,
# simulations, random initialization, and other
# machine learning operations.
```

# 2. Random Seed

A random seed is a starting value used by a random number generator.

In NumPy, we use `np.random.seed()` to set the random seed.

When the same seed is used, NumPy generates the same sequence of random numbers.

This makes our random results reproducible.

In AI/ML, reproducibility is important because we may need to get the same results when testing, debugging, comparing models, or sharing experiments.

# Simple / AI-ML Example

Suppose we generate random values for an ML experiment.

Without setting a seed, the random values may be different every time we run the program.

If we use:

`np.random.seed(42)`

and generate the random values again with the same seed, we will get the same values.

This helps us reproduce the same experiment and compare results fairly.

In AI/ML, random seeds are commonly used during model training, data splitting, and experimentation.

```python
import numpy as np

# Set the random seed
np.random.seed(42)

# Generate random numbers
random_numbers_1 = np.random.rand(5)

# Reset the same random seed
np.random.seed(42)

# Generate random numbers again
random_numbers_2 = np.random.rand(5)

print("First Result:", random_numbers_1)
print("Second Result:", random_numbers_2)

# Check whether both results are the same
print("Results are same:", np.array_equal(random_numbers_1, random_numbers_2))

# Explanation:
# np.random.seed(42) fixes the starting point of the
# random number generator.
#
# Using the same seed produces the same sequence of
# random numbers.
#
# Expected Output:
# First Result: [0.37454012 0.95071431 0.73199394 0.59865848 0.15601864]
# Second Result: [0.37454012 0.95071431 0.73199394 0.59865848 0.15601864]
# Results are same: True
#
# AI/ML Use:
# Random seeds help make ML experiments reproducible,
# making it easier to debug, test, and compare models.
```

# 3. Random Integers

Random integers are randomly generated whole numbers.

In NumPy, we can use `np.random.randint()` to generate random integers within a specified range.

The lower limit is included, while the upper limit is excluded.

For example:

`np.random.randint(1, 10)`

can generate an integer from 1 to 9.

Random integers are useful for simulations, testing, sampling, and creating random datasets.

In AI/ML, they can be used for generating sample data, random indexing, and simulation tasks.

# Simple / AI-ML Example

Suppose we want to randomly select five student IDs from the range 1 to 100.

We can use `np.random.randint()` to generate random integer values.

For example:

[23, 67, 12, 89, 45]

These values can represent randomly selected IDs.

In AI/ML and Data Analytics, random integers can be useful for creating sample data, selecting random positions, and running simulations.

```python
import numpy as np

# Set a seed so that the result is reproducible
np.random.seed(42)

# Generate 5 random integers from 1 to 100
student_ids = np.random.randint(1, 101, size=5)

print("Random Student IDs:", student_ids)

# Explanation:
# np.random.randint() generates random integer values.
# 1 is the lower limit and 101 is the upper limit.
# The upper limit is excluded, so values can range from 1 to 100.
# size=5 means that 5 random integers are generated.
#
# Expected Output:
# Random Student IDs: [52 93 15 72 61]
#
# AI/ML Use:
# Random integers can be used for sampling,
# random indexing, test data generation, and simulations.
```

# 4. Random Choice

Random choice is used to randomly select values from an array or a collection of values.

In NumPy, we use `np.random.choice()` to randomly select one or more values.

It can select a single value or multiple values from the given data.

Random choice is useful for sampling data without manually selecting values.

In AI/ML, random choice can be used for creating random samples, selecting data points, and simulation tasks.

# Simple / AI-ML Example

Suppose we have four product categories:

["Laptop", "Mobile", "Tablet", "Watch"]

If we want to randomly select two products for a sample experiment, we can use `np.random.choice()`.

For example, the selected products could be:

["Mobile", "Watch"]

The selection can be different each time if a seed is not used.

In AI/ML and Data Analytics, random choice can be useful for sampling observations from a dataset.

```python
import numpy as np

# Set a seed to make the random selection reproducible
np.random.seed(42)

# Create an array of product categories
products = np.array(["Laptop", "Mobile", "Tablet", "Watch"])

# Randomly select 2 products
selected_products = np.random.choice(products, size=2, replace=False)

print("Products:", products)
print("Randomly Selected Products:", selected_products)

# Explanation:
# np.random.choice() randomly selects values from an array.
# size=2 means that 2 values are selected.
# replace=False means the same value cannot be selected twice.
#
# Expected Output:
# Products: ['Laptop' 'Mobile' 'Tablet' 'Watch']
# Randomly Selected Products: ['Tablet' 'Watch']
#
# AI/ML Use:
# Random choice can be used for random sampling,
# selecting observations, and creating sample datasets
# for experiments and analysis.
```

# 5. Shuffle

Shuffle is used to randomly change the order of elements in an array.

In NumPy, we use `np.random.shuffle()` to shuffle the elements of an array.

The shuffle operation changes the original array directly.

Shuffling is useful when we want to remove the existing order of data.

In AI/ML, shuffling is commonly used before training a model so that the model does not learn from the original order of the dataset.

# Simple / AI-ML Example

Suppose we have student records in a particular order:

[1, 2, 3, 4, 5]

If we shuffle the records, the order may become:

[3, 1, 5, 2, 4]

The values remain the same, but their order changes.

In AI/ML, shuffling a dataset before training helps prevent the model from being affected by the original ordering of the data.

```python
import numpy as np

# Create an array of student IDs
student_ids = np.array([1, 2, 3, 4, 5])

# Set a seed so that the shuffle is reproducible
np.random.seed(42)

# Shuffle the array in place
np.random.shuffle(student_ids)

print("Shuffled Student IDs:", student_ids)

# Explanation:
# np.random.shuffle() randomly changes the order
# of elements in the original array.
#
# The original array itself is modified.
#
# Expected Output:
# Shuffled Student IDs: [2 5 3 1 4]
#
# AI/ML Use:
# Shuffling is commonly used before model training
# to randomize the order of training data and reduce
# the effect of any existing order in the dataset.
```

# 6. Permutation

Permutation generates a randomly rearranged copy of an array.

In NumPy, we use `np.random.permutation()` to create a random permutation.

Unlike `np.random.shuffle()`, permutation does **not change the original array**. It returns a new shuffled array.

This is useful when we want a randomized order while keeping the original data unchanged.

In AI/ML, permutation can be used for creating randomized datasets, testing, and evaluating how changes in data order affect a model.

# Simple / AI-ML Example

Suppose we have student IDs:

[1, 2, 3, 4, 5]

Using permutation, we can create a randomly rearranged version:

[2, 5, 1, 4, 3]

The original array remains unchanged.

In AI/ML, permutation can be useful when we need a randomized version of data for experiments while preserving the original dataset.

```python
import numpy as np

# Create an array of student IDs
student_ids = np.array([1, 2, 3, 4, 5])

# Set a seed so that the result is reproducible
np.random.seed(42)

# Create a random permutation of the array
permuted_ids = np.random.permutation(student_ids)

print("Original Student IDs:", student_ids)
print("Permuted Student IDs:", permuted_ids)

# Explanation:
# np.random.permutation() returns a randomly rearranged
# copy of the array.
#
# The original array is not changed.
#
# Expected Output:
# Original Student IDs: [1 2 3 4 5]
# Permuted Student IDs: [2 5 3 1 4]
#
# AI/ML Use:
# Permutation is useful for randomizing data,
# creating experiments, and testing the effect of
# changing the order of observations.
```

# 7. Random Distributions

A random distribution describes how randomly generated values are spread across a range of possible values.

NumPy provides different random distribution functions through `np.random`.

Some commonly used distributions are:

- `np.random.normal()` – generates values following a normal distribution.
- `np.random.uniform()` – generates values with a uniform distribution.
- `np.random.binomial()` – generates values based on binomial probability.

For example, a normal distribution is commonly used when values are concentrated around a mean.

In AI/ML, random distributions are useful for generating sample data, simulations, model initialization, and understanding probability-based data.

# Simple / AI-ML Example

Suppose the heights of students approximately follow a normal distribution.

We can generate random height values using:

`np.random.normal()`

For example, we can specify:

- Mean = 170 cm
- Standard deviation = 5 cm
- Number of values = 10

The generated values will generally be centered around 170 cm, with most values closer to the mean.

In AI/ML, random distributions can be used to simulate data and initialize model parameters.

```python
import numpy as np

# Set a seed so that the random values are reproducible
np.random.seed(42)

# Generate 10 random heights using a normal distribution
# Mean = 170 cm and Standard Deviation = 5 cm
heights = np.random.normal(loc=170, scale=5, size=10)

print("Random Heights:")
print(heights)

# Explanation:
# np.random.normal() generates random values following
# a normal distribution.
#
# loc=170 represents the mean.
# scale=5 represents the standard deviation.
# size=10 means that 10 random values are generated.
#
# Expected Output:
# The output will contain 10 random height values
# centered approximately around 170 cm.
#
# AI/ML Use:
# Random distributions are useful for data simulation,
# statistical experiments, and initializing values
# used in machine learning models.
```
