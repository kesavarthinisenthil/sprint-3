# groupby()

## Understand the Concept

In Pandas, `groupby()` is used to group rows based on one or more columns.

After grouping the data, we can perform calculations such as sum, mean, count, minimum, and maximum on each group.

### Why do we use it?

- To organize data into groups.
- To compare different groups.
- To calculate summaries for each group.
- To analyze patterns in data.
- To perform group-level data analysis.

### AI/ML Use

In AI/ML and Data Analytics, `groupby()` can be used to understand patterns within different groups.

For example, we can group customer data by city and calculate the average purchase amount for each city.

## Demonstrate the Concept

### Real-World Example

Suppose we have sales data containing:

- `City`
- `Product`
- `Sales`

We can group the data by `City` and calculate the total sales for each city.

This helps a business understand which cities generate more sales.

### AI/ML Example

Suppose we have customer data containing customer purchases from different cities.

We can use `groupby()` to calculate the average purchase amount for each city.

This can help identify customer spending patterns and can be useful during data analysis and feature engineering.

```python
import pandas as pd

# Create a sample sales DataFrame
df = pd.DataFrame({
    "City": ["Chennai", "Salem", "Chennai", "Salem", "Coimbatore"],
    "Sales": [5000, 3000, 7000, 4000, 6000]
})

# Display the original DataFrame
print("Original DataFrame:")
print(df)

# Group the data by City and calculate total sales
# groupby() groups rows with the same city.
# sum() calculates the total sales for each city.
city_sales = df.groupby("City")["Sales"].sum()

# Display the result
print("\nTotal Sales by City:")
print(city_sales)

# Expected output:
# Chennai     12000
# Coimbatore   6000
# Salem        7000

# AI/ML / Data Analytics Use:
# groupby() helps analyze patterns and summarize data
# for different groups during data analysis and feature engineering.
```

# agg()

## Understand the Concept

In Pandas, `agg()` is used to perform multiple aggregation operations on grouped data.

We can use functions such as `sum()`, `mean()`, `min()`, `max()`, and `count()` together.

### Why do we use it?

- To perform multiple calculations at the same time.
- To summarize grouped data.
- To calculate different statistics for different columns.
- To make data analysis easier and more efficient.

### AI/ML Use

In AI/ML and Data Analytics, `agg()` can be used to generate multiple summary statistics from grouped data.

For example, we can calculate the total sales, average sales, and maximum sales for each city.

## Demonstrate the Concept

### Real-World Example

Suppose we have sales data from different cities.

We want to know:

- Total sales for each city.
- Average sales for each city.
- Maximum sales for each city.

Instead of calculating each value separately, we can use `agg()` to perform all these calculations together.

### AI/ML Example

Suppose we have customer purchase data grouped by city.

Using `agg()`, we can calculate multiple statistics such as average, minimum, and maximum purchase amounts for each city.

These statistics can help us understand customer behavior and create useful features for machine learning.

```python
import pandas as pd

# Create a sample sales DataFrame
df = pd.DataFrame({
    "City": ["Chennai", "Salem", "Chennai", "Salem", "Coimbatore"],
    "Sales": [5000, 3000, 7000, 4000, 6000]
})

# Display the original DataFrame
print("Original DataFrame:")
print(df)

# Group the data by City and perform multiple calculations
# agg() allows us to apply multiple aggregation functions at once.
city_summary = df.groupby("City")["Sales"].agg(["sum", "mean", "max"])

# Display the result
print("\nSales Summary by City:")
print(city_summary)

# Expected output:
# The result will contain the total, average, and maximum
# sales for each city.

# AI/ML / Data Analytics Use:
# agg() helps generate multiple statistical summaries
# that can be useful for data analysis and feature engineering.
```

# transform()

## Understand the Concept

In Pandas, `transform()` is used to perform a calculation on grouped data while keeping the same number of rows as the original DataFrame.

Unlike `agg()`, which reduces the data into summary rows, `transform()` returns a value for every original row.

### Why do we use it?

- To perform calculations within groups.
- To keep the original DataFrame size.
- To create new columns based on group-level calculations.
- To compare individual values with their group.

### AI/ML Use

In AI/ML and Data Analytics, `transform()` can be used to create new features based on group-level statistics.

For example, we can calculate the average sales for each city and add that average to every row belonging to that city.

## Demonstrate the Concept

### Real-World Example

Suppose we have sales data from different cities.

We want to find the average sales of each city and show that average alongside every individual sales record.

`transform()` allows us to calculate the city average while keeping every original row.

### AI/ML Example

Suppose customer purchase data is grouped by city.

We can calculate the average purchase amount for each city and add it as a new feature to every customer record.

This helps compare an individual customer's purchase with the average purchase of their city.

```python
import pandas as pd

# Create a sample sales DataFrame
df = pd.DataFrame({
    "City": ["Chennai", "Salem", "Chennai", "Salem", "Coimbatore"],
    "Sales": [5000, 3000, 7000, 4000, 6000]
})

# Display the original DataFrame
print("Original DataFrame:")
print(df)

# Calculate the average sales for each city
# transform() returns the group average for every original row.
df["City_Average_Sales"] = df.groupby("City")["Sales"].transform("mean")

# Display the updated DataFrame
print("\nDataFrame after using transform():")
print(df)

# Expected output:
# Each row will contain the average sales of its corresponding city.
# Chennai average = 6000
# Salem average = 3500
# Coimbatore average = 6000

# AI/ML / Data Analytics Use:
# transform() is useful for creating group-based features
# while keeping the original number of rows in the dataset.
```

# filter()

## Understand the Concept

In Pandas, `filter()` can be used with grouped data to keep or remove entire groups based on a condition.

It is useful when we want to select groups that satisfy a specific requirement.

### Why do we use it?

- To select groups based on a condition.
- To remove groups that do not meet a requirement.
- To focus on specific groups during analysis.
- To perform group-level filtering.

### AI/ML Use

In AI/ML and Data Analytics, `filter()` can be used to select groups that meet a particular condition.

For example, we can keep only cities whose total sales are greater than a specific amount.

## Demonstrate the Concept

### Real-World Example

Suppose we have sales data from different cities.

A business may want to analyze only cities where the total sales are greater than 10,000.

We can use `groupby()` with `filter()` to keep only those cities.

### AI/ML Example

Suppose we have customer data grouped by city.

We may want to analyze only cities that have enough customer records for reliable analysis.

Using `filter()`, we can keep groups that satisfy the required condition.

```python
import pandas as pd

# Create a sample sales DataFrame
df = pd.DataFrame({
    "City": ["Chennai", "Salem", "Chennai", "Salem", "Coimbatore"],
    "Sales": [5000, 3000, 7000, 4000, 6000]
})

# Display the original DataFrame
print("Original DataFrame:")
print(df)

# Keep only cities whose total sales are greater than 10000
# groupby() creates groups based on City.
# filter() keeps only the groups that satisfy the condition.
filtered_df = df.groupby("City").filter(
    lambda group: group["Sales"].sum() > 10000
)

# Display the filtered DataFrame
print("\nDataFrame after using filter():")
print(filtered_df)

# Expected output:
# Chennai has total sales of 12000, so its rows are kept.
# Salem has total sales of 7000, so its rows are removed.
# Coimbatore has total sales of 6000, so its row is removed.

# AI/ML / Data Analytics Use:
# filter() helps select relevant groups for analysis
# based on group-level conditions.
```

# Pivot Tables

## Understand the Concept

In Pandas, a pivot table is used to summarize and organize data based on one or more columns.

It allows us to calculate values such as sum, mean, count, minimum, and maximum for different groups.

Pivot tables make large datasets easier to analyze and understand.

### Why do we use Pivot Tables?

- To summarize large amounts of data.
- To compare values across different categories.
- To identify patterns and trends.
- To organize data in a meaningful table.
- To perform quick data analysis.

### AI/ML Use

In AI/ML and Data Analytics, pivot tables can be used to summarize datasets and understand relationships between different features.

For example, we can summarize the average sales for different products across different cities.

## Demonstrate the Concept

### Real-World Example

Suppose a company has sales data containing:

- `City`
- `Product`
- `Sales`

A pivot table can show the total sales of each product for each city.

This helps the business compare product performance across different cities.

### AI/ML Example

Suppose we have customer purchase data.

A pivot table can summarize the average purchase amount for different customer categories and locations.

This can help us identify patterns in customer behavior during data analysis.

```python
import pandas as pd

# Create a sample sales DataFrame
df = pd.DataFrame({
    "City": ["Chennai", "Chennai", "Salem", "Salem", "Coimbatore"],
    "Product": ["Laptop", "Phone", "Laptop", "Phone", "Laptop"],
    "Sales": [50000, 30000, 45000, 25000, 55000]
})

# Display the original DataFrame
print("Original DataFrame:")
print(df)

# Create a pivot table
# pivot_table() summarizes data based on selected rows and columns.
pivot = pd.pivot_table(
    df,
    values="Sales",
    index="City",
    columns="Product",
    aggfunc="sum",
    fill_value=0
)

# Display the pivot table
print("\nPivot Table:")
print(pivot)

# Expected output:
# The table will show total sales for each product
# across different cities.

# AI/ML / Data Analytics Use:
# Pivot tables help summarize and compare data,
# making patterns and relationships easier to understand.
```

# Crosstab

## Understand the Concept

In Pandas, `crosstab()` is used to create a frequency table that shows the relationship between two or more categorical variables.

It counts how often each combination of categories occurs.

### Why do we use it?

- To understand relationships between categorical variables.
- To count occurrences of categories.
- To compare different categories.
- To summarize categorical data.
- To identify patterns in data.

### AI/ML Use

In AI/ML and Data Analytics, `crosstab()` can be used to explore relationships between categorical features.

For example, we can analyze the relationship between `Gender` and `Purchased` to understand customer behavior before building a machine learning model.

## Demonstrate the Concept

### Real-World Example

Suppose a company has customer data containing:

- `Gender`
- `Purchased`

We can use a crosstab to count how many customers from each gender purchased a product.

This helps the business understand purchasing patterns.

### AI/ML Example

Before building a machine learning model, we can use `crosstab()` to explore relationships between categorical features and a target variable.

For example, we can compare `Gender` with `Purchased` to understand whether purchasing behavior differs between groups.

```python
import pandas as pd

# Create a sample customer DataFrame
df = pd.DataFrame({
    "Gender": ["Male", "Female", "Male", "Female", "Male", "Female"],
    "Purchased": ["Yes", "Yes", "No", "Yes", "No", "No"]
})

# Display the original DataFrame
print("Original DataFrame:")
print(df)

# Create a crosstab
# crosstab() counts the number of occurrences
# between the Gender and Purchased columns.
table = pd.crosstab(df["Gender"], df["Purchased"])

# Display the crosstab
print("\nCrosstab:")
print(table)

# Expected output:
# The table will show the number of Yes and No purchases
# for each gender.

# AI/ML / Data Analytics Use:
# crosstab() helps explore relationships between categorical
# variables during exploratory data analysis.
```
