# concat()

## Understand the Concept

In Pandas, `concat()` is used to combine two or more DataFrames along rows or columns.

By default, `concat()` combines DataFrames vertically, meaning one DataFrame is placed below another.

We can also use `axis=1` to combine DataFrames side by side.

### Why do we use it?

- To combine multiple DataFrames.
- To append rows from different datasets.
- To combine data with similar structures.
- To combine data vertically or horizontally.

### AI/ML Use

In AI/ML projects, data may be collected from different files or sources.

We can use `concat()` to combine these datasets into one DataFrame before performing data analysis and preprocessing.

## Demonstrate the Concept

### Real-World Example

Suppose a company stores sales data in separate DataFrames for January and February.

Both DataFrames have the same columns.

We can use `concat()` to combine the two DataFrames and create one dataset containing sales from both months.

### AI/ML Example

Suppose training data is collected from different sources but has the same structure.

We can use `concat()` to combine the datasets into one DataFrame before preprocessing and model training.

```python
import pandas as pd

# Create January sales DataFrame
january = pd.DataFrame({
    "Name": ["Arun", "Priya"],
    "Sales": [5000, 7000]
})

# Create February sales DataFrame
february = pd.DataFrame({
    "Name": ["Kumar", "Divya"],
    "Sales": [6000, 8000]
})

# Display the original DataFrames
print("January Data:")
print(january)

print("\nFebruary Data:")
print(february)

# Combine both DataFrames row-wise
# concat() is used to combine DataFrames.
# ignore_index=True creates a new continuous index.
combined = pd.concat([january, february], ignore_index=True)

# Display the combined DataFrame
print("\nCombined Data:")
print(combined)

# Expected output:
# The rows from February will be added below the January rows.

# AI/ML / Data Analytics Use:
# concat() is useful for combining datasets collected
# from different files, periods, or data sources.
```

# merge()

## Understand the Concept

In Pandas, `merge()` is used to combine two DataFrames based on a common column or key.

It works similarly to JOIN operations in SQL.

We can use different types of joins such as inner, left, right, and outer.

### Why do we use it?

- To combine related datasets.
- To connect information from different DataFrames.
- To match rows using a common column.
- To perform database-like joins in Pandas.

### AI/ML Use

In AI/ML projects, useful information may be stored in different datasets.

For example, customer details may be stored in one DataFrame and purchase details in another.

We can use `merge()` to combine them using a common `Customer_ID`.

## Demonstrate the Concept

### Real-World Example

Suppose we have two datasets:

- Customer details containing `Customer_ID` and `Name`.
- Purchase details containing `Customer_ID` and `Amount`.

Both datasets have `Customer_ID` in common.

We can use `merge()` to combine the customer information with their purchase information.

### AI/ML Example

In an ML project, customer demographic information and transaction information may be stored separately.

We can merge these datasets using `Customer_ID` to create a complete dataset for analysis and model training.

```python
import pandas as pd

# Create a customer DataFrame
customers = pd.DataFrame({
    "Customer_ID": [1, 2, 3],
    "Name": ["Arun", "Priya", "Kumar"]
})

# Create a purchase DataFrame
purchases = pd.DataFrame({
    "Customer_ID": [1, 2, 3],
    "Purchase_Amount": [5000, 7500, 6200]
})

# Display the original DataFrames
print("Customers:")
print(customers)

print("\nPurchases:")
print(purchases)

# Merge the two DataFrames using Customer_ID
# merge() combines related data using a common column.
merged_df = pd.merge(customers, purchases, on="Customer_ID")

# Display the merged DataFrame
print("\nMerged DataFrame:")
print(merged_df)

# Expected output:
# The Name and Purchase_Amount columns will be combined
# using Customer_ID as the common key.

# AI/ML / Data Analytics Use:
# merge() is useful for combining related datasets
# before analysis, preprocessing, and machine learning.
```

# join()

## Understand the Concept

In Pandas, `join()` is used to combine two DataFrames based on their index.

It is useful when related information is stored in different DataFrames and they have matching index values.

### Why do we use it?

- To combine DataFrames using their index.
- To add related columns from another DataFrame.
- To combine data quickly when indexes are aligned.
- To work with datasets that use meaningful index values.

### AI/ML Use

In AI/ML and Data Analytics, `join()` can be used to combine related datasets that share the same index.

For example, customer information and customer purchase data can be joined when both DataFrames use the same customer ID as their index.

## Demonstrate the Concept

### Real-World Example

Suppose we have two DataFrames:

- One contains customer names.
- Another contains customer purchase amounts.

Both DataFrames use `Customer_ID` as their index.

We can use `join()` to combine the purchase information with the customer information.

### AI/ML Example

Suppose customer demographic features and transaction features are stored in separate DataFrames with the same customer ID as the index.

We can use `join()` to combine these features into a single dataset for analysis or machine learning.

```python
import pandas as pd

# Create a customer DataFrame with Customer_ID as the index
customers = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar"]
}, index=[1, 2, 3])

# Create a purchase DataFrame with the same index
purchases = pd.DataFrame({
    "Purchase_Amount": [5000, 7500, 6200]
}, index=[1, 2, 3])

# Display the original DataFrames
print("Customers:")
print(customers)

print("\nPurchases:")
print(purchases)

# Join the two DataFrames using their index
# join() combines columns from DataFrames based on matching indexes.
joined_df = customers.join(purchases)

# Display the joined DataFrame
print("\nJoined DataFrame:")
print(joined_df)

# Expected output:
# The Name and Purchase_Amount columns will be combined
# based on their matching index values.

# AI/ML / Data Analytics Use:
# join() is useful for combining related features
# when DataFrames have the same index.
```

# append()

## Understand the Concept

In older versions of Pandas, `append()` was used to add rows from one DataFrame to another DataFrame.

However, `append()` has been deprecated and removed in newer versions of Pandas.

The recommended alternative is `concat()`.

### Why do we use it?

- To add rows to a DataFrame.
- To combine two DataFrames vertically.
- To add new records to existing data.

### AI/ML Use

In AI/ML projects, we may need to add new records to an existing dataset.

Since `append()` is no longer recommended, we use `concat()` as the modern alternative for combining DataFrames.

## Demonstrate the Concept

### Real-World Example

Suppose a company has sales data for January and receives new sales data for February.

Earlier, `append()` could be used to add the February records to the January DataFrame.

In modern Pandas, we use `concat()` instead.

### AI/ML Example

Suppose new training data is collected from another source.

We can use `concat()` to combine the existing dataset with the new records.

This is the modern alternative to the deprecated `append()` method.

```python
import pandas as pd

# Create the existing DataFrame
df1 = pd.DataFrame({
    "Name": ["Arun", "Priya"],
    "Sales": [5000, 7000]
})

# Create a new DataFrame
df2 = pd.DataFrame({
    "Name": ["Kumar", "Divya"],
    "Sales": [6000, 8000]
})

# Display the original DataFrames
print("Existing Data:")
print(df1)

print("\nNew Data:")
print(df2)

# append() is deprecated and removed in modern Pandas.
# Therefore, concat() is used as the recommended alternative.
combined = pd.concat([df1, df2], ignore_index=True)

# Display the combined DataFrame
print("\nCombined Data:")
print(combined)

# Expected output:
# The rows from df2 will be added below the rows from df1.

# AI/ML / Data Analytics Use:
# concat() is used to combine new records with existing
# data and is the modern alternative to append().
```

# combine_first()

## Understand the Concept

In Pandas, `combine_first()` is used to fill missing values in one DataFrame using values from another DataFrame.

It keeps the existing values in the first DataFrame and uses values from the second DataFrame only where the first DataFrame has missing values.

### Why do we use it?

- To fill missing values using another DataFrame.
- To combine information from two datasets.
- To preserve existing values.
- To handle missing data efficiently.

### AI/ML Use

In AI/ML projects, information may be available from multiple data sources.

If one dataset has missing values, `combine_first()` can use values from another dataset to fill those missing values before data analysis or model training.

## Demonstrate the Concept

### Real-World Example

Suppose we have two customer datasets.

The first dataset contains customer names and phone numbers, but some phone numbers are missing.

The second dataset contains some of the missing phone numbers.

We can use `combine_first()` to fill the missing values in the first dataset using the second dataset.

### AI/ML Example

Suppose two datasets contain customer features, but some values are missing in the first dataset.

We can use another dataset as a backup source to fill those missing values.

This can help create a more complete dataset for analysis and machine learning.

```python
import pandas as pd

# Create the first DataFrame with some missing values
df1 = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar"],
    "Age": [25, None, 28]
}, index=[1, 2, 3])

# Create the second DataFrame containing values for missing data
df2 = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar"],
    "Age": [25, 30, 28]
}, index=[1, 2, 3])

# Display the original DataFrames
print("First DataFrame:")
print(df1)

print("\nSecond DataFrame:")
print(df2)

# Fill missing values in df1 using values from df2
# combine_first() keeps existing values in df1
# and uses df2 values only where df1 has missing values.
combined = df1.combine_first(df2)

# Display the combined DataFrame
print("\nCombined DataFrame:")
print(combined)

# Expected output:
# The missing Age value for Priya will be filled with 30
# from the second DataFrame.

# AI/ML / Data Analytics Use:
# combine_first() can help combine data sources and
# fill missing values during data preprocessing.
```
