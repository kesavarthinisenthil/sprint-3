# isnull()

## Understand the Concept

In Pandas, `isnull()` is used to identify missing or empty values in a DataFrame or Series.

It returns `True` when a value is missing and `False` when a value is present.

### Why do we use it?

- To identify missing values.
- To check which columns contain missing data.
- To understand the quality of a dataset.
- To prepare data for cleaning and analysis.

### AI/ML Use

In AI/ML projects, datasets may contain missing values.

Using `isnull()` helps identify missing data before preprocessing and model training.

## Demonstrate the Concept

### Real-World Example

Suppose a company has customer information containing Name, Age, and City.

Some customers may not have provided their age.

We can use `isnull()` to identify which records have missing age values.

### AI/ML Example

Suppose a machine learning dataset contains missing values in important features such as Age or Salary.

We can use `isnull()` to find these missing values and decide how to handle them before training the model.

```python
import pandas as pd

# Create a DataFrame containing missing values
data = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar"],
    "Age": [25, None, 30],
    "City": ["Chennai", "Coimbatore", None]
})

# Display the original data
print("Original Data:")
print(data)

# Check for missing values
# isnull() returns True for missing values
# and False for non-missing values.
print("\nMissing Values:")
print(data.isnull())

# Count the missing values in each column
print("\nMissing Value Count:")
print(data.isnull().sum())

# Expected output:
# Age has 1 missing value.
# City has 1 missing value.

# AI/ML / Data Analytics Use:
# isnull() helps identify missing values before
# data cleaning, preprocessing, and model training.
```

# notnull()

## Understand the Concept

In Pandas, `notnull()` is used to identify values that are not missing.

It returns `True` when a value is present and `False` when a value is missing.

### Why do we use it?

- To identify available values.
- To filter rows containing valid data.
- To check which values are present.
- To work with complete records.

### AI/ML Use

In AI/ML projects, `notnull()` can be used to select records that contain valid values before performing data analysis or model training.

## Demonstrate the Concept

### Real-World Example

Suppose a customer dataset contains customer names and phone numbers.

Some customers may not have provided their phone number.

We can use `notnull()` to find the customers who have provided a phone number.

### AI/ML Example

Suppose a machine learning dataset contains missing values in an important feature.

We can use `notnull()` to identify rows where that feature contains a valid value and use those records for further analysis or preprocessing.

```python
import pandas as pd

# Create a DataFrame containing missing values
data = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar"],
    "Age": [25, None, 30],
    "City": ["Chennai", "Coimbatore", None]
})

# Display the original data
print("Original Data:")
print(data)

# Check which values are not missing
# notnull() returns True for available values
# and False for missing values.
print("\nNon-Missing Values:")
print(data.notnull())

# Select rows where Age is available
valid_age = data[data["Age"].notnull()]

# Display rows with available Age values
print("\nRows with Available Age:")
print(valid_age)

# Expected output:
# Rows where Age is 25 and 30 will be selected.
# The row with missing Age will not be selected.

# AI/ML / Data Analytics Use:
# notnull() helps identify and filter records with
# valid data during data cleaning and preprocessing.
```

# dropna()

## Understand the Concept

In Pandas, `dropna()` is used to remove rows or columns that contain missing values.

It is useful when missing data cannot be filled reliably or when we want to work only with complete records.

### Why do we use it?

- To remove missing values.
- To create a cleaner dataset.
- To work with complete records.
- To prepare data for analysis and machine learning.

### AI/ML Use

In AI/ML projects, rows with missing values may sometimes be removed when the amount of missing data is small or when the missing records are not useful for model training.

## Demonstrate the Concept

### Real-World Example

Suppose a company has customer records where some customers have not provided their age.

If only a few records have missing age values, the company can remove those records before analysis.

We can use `dropna()` to remove rows containing missing values.

### AI/ML Example

Suppose a machine learning dataset contains a small number of incomplete records.

If removing those records does not significantly reduce the dataset, `dropna()` can be used to remove them before model training.

```python
import pandas as pd

# Create a DataFrame containing missing values
data = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar", "Divya"],
    "Age": [25, None, 30, 28],
    "City": ["Chennai", "Coimbatore", None, "Salem"]
})

# Display the original data
print("Original Data:")
print(data)

# Remove rows containing any missing value
# dropna() removes rows with missing values.
clean_data = data.dropna()

# Display the cleaned data
print("\nAfter using dropna():")
print(clean_data)

# Expected output:
# Rows containing missing Age or City values will be removed.

# AI/ML / Data Analytics Use:
# dropna() can be used to remove incomplete records
# when the amount of missing data is small.
```

# fillna()

## Understand the Concept

In Pandas, `fillna()` is used to replace missing values with a specified value.

Instead of removing rows containing missing data, we can use `fillna()` to fill those missing values.

### Why do we use it?

- To replace missing values.
- To keep useful records in the dataset.
- To avoid losing data by removing rows.
- To prepare complete data for analysis and machine learning.

### AI/ML Use

In AI/ML projects, missing values can affect model performance.

We can use `fillna()` to replace missing values with suitable values such as the mean, median, mode, or a fixed value during data preprocessing.

## Demonstrate the Concept

### Real-World Example

Suppose a customer dataset contains missing Age values.

Instead of removing those customer records, we can replace the missing Age with the average age of the available customers.

This allows us to keep the records in the dataset.

### AI/ML Example

Suppose a machine learning dataset contains missing values in a numerical feature.

We can use `fillna()` with the mean or median to fill the missing values before training the model.

This helps maintain the size of the dataset.

```python
import pandas as pd

# Create a DataFrame containing missing Age values
data = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar", "Divya"],
    "Age": [25, None, 30, None],
    "City": ["Chennai", "Coimbatore", "Salem", "Madurai"]
})

# Display the original data
print("Original Data:")
print(data)

# Calculate the mean Age
mean_age = data["Age"].mean()

# Fill missing Age values with the mean Age
# fillna() replaces missing values with the specified value.
data["Age"] = data["Age"].fillna(mean_age)

# Display the updated data
print("\nAfter using fillna():")
print(data)

# Expected output:
# Missing Age values will be replaced with the mean Age.

# AI/ML / Data Analytics Use:
# fillna() is commonly used during data preprocessing
# to handle missing values without removing records.
```

# interpolate()

## Understand the Concept

In Pandas, `interpolate()` is used to estimate and fill missing values based on the values before and after the missing value.

It is especially useful for numerical data that follows a sequence or trend.

### Why do we use it?

- To fill missing numerical values.
- To estimate values between known values.
- To preserve the pattern or trend in the data.
- To handle missing values without simply using the mean or median.

### AI/ML Use

In AI/ML projects, `interpolate()` can be useful for time-series data such as sensor readings, temperature measurements, stock data, or machine monitoring data where values usually follow a continuous pattern.

## Demonstrate the Concept

### Real-World Example

Suppose a machine records its temperature every hour:

- 20°C
- 22°C
- Missing
- 26°C

We can use interpolation to estimate the missing value between 22°C and 26°C.

The estimated value would be 24°C.

### AI/ML Example

Suppose a sensor records temperature continuously, but one reading is missing.

Instead of removing the record or replacing it with the overall average, we can use interpolation to estimate the missing value based on nearby readings.

This helps preserve the time-series pattern.

```python
import pandas as pd

# Create a DataFrame with a missing temperature value
data = pd.DataFrame({
    "Hour": [1, 2, 3, 4, 5],
    "Temperature": [20, 22, None, 26, 28]
})

# Display the original data
print("Original Data:")
print(data)

# Fill the missing temperature using interpolation
# interpolate() estimates the missing value using
# surrounding values.
data["Temperature"] = data["Temperature"].interpolate()

# Display the updated data
print("\nAfter using interpolate():")
print(data)

# Expected output:
# The missing temperature at Hour 3 will be estimated
# as 24.0 based on the values around it.

# AI/ML / Data Analytics Use:
# interpolate() is useful for filling missing values
# in continuous and time-series data.
```

# Missing Value Handling Strategies

## Understand the Concept

Missing value handling strategies are different methods used to deal with missing data in a dataset.

Common strategies include removing missing values, filling them with statistical values such as mean, median, or mode, and using interpolation for continuous data.

### Why do we use it?

- To prepare clean and complete data.
- To avoid errors during data analysis.
- To prevent machine learning models from being affected by missing values.
- To choose the most suitable method based on the dataset.

### AI/ML Use

In AI/ML projects, the best missing value strategy depends on the type of data and the reason for the missing values.

For example:

- Mean or median can be used for numerical data.
- Mode can be used for categorical data.
- `dropna()` can be used when only a few records are missing.
- `interpolate()` can be useful for time-series data.

## Demonstrate the Concept

### Real-World Example

Suppose a customer dataset contains missing Age values.

We can use different strategies depending on the situation:

- Remove rows using `dropna()`.
- Fill missing Age using the mean or median with `fillna()`.
- Fill missing categorical values using the mode.
- Use `interpolate()` when the data follows a continuous pattern.

The choice depends on the dataset and the amount of missing data.

### AI/ML Example

Suppose a machine learning dataset contains missing values in customer information.

We can use the median for numerical features because it is less affected by extreme values.

For categorical features, we can use the mode.

Choosing an appropriate strategy helps create better-quality data for model training.

```python
import pandas as pd

# Create a DataFrame containing missing values
data = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar", "Divya"],
    "Age": [25, None, 30, None],
    "City": ["Chennai", "Coimbatore", None, "Salem"]
})

# Display the original data
print("Original Data:")
print(data)

# Fill missing Age values using the median
# Median is useful for numerical data and is less affected
# by extreme values.
median_age = data["Age"].median()
data["Age"] = data["Age"].fillna(median_age)

# Fill the missing City value using the mode
# Mode represents the most frequently occurring value.
mode_city = data["City"].mode()[0]
data["City"] = data["City"].fillna(mode_city)

# Display the cleaned data
print("\nAfter Handling Missing Values:")
print(data)

# Expected output:
# Missing Age values will be replaced with the median.
# The missing City value will be replaced with the mode.

# AI/ML / Data Analytics Use:
# Choosing the correct missing value strategy is an
# important part of data preprocessing before model training.
```
