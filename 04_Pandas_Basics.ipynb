# What is Pandas?

## Understand the Concept

Pandas is a Python library used for **working with and analyzing structured data**.

It provides two main data structures:
- **Series** – one-dimensional data
- **DataFrame** – two-dimensional data in rows and columns

Pandas helps us to easily **read, clean, filter, manipulate, and analyze data**.

In AI/ML, Pandas is mainly used during **data preprocessing and data analysis** before building machine learning models.

## Demonstrate the Concept

### Real-World Example

Consider a student dataset containing names, ages, and marks.

Instead of storing and handling each value separately, Pandas allows us to organize the information into a table using a DataFrame.

For example:

Name | Age | Marks
Arun | 21 | 85
Ravi | 22 | 90
Priya | 21 | 88

We can then easily analyze this data, such as finding marks, filtering students, or calculating the average.

### AI/ML Example

Suppose we have a dataset containing customer information.

Using Pandas, we can load the dataset, check the data, find missing values, filter required records, and prepare the data for a machine learning model.

So, Pandas makes **data handling and analysis easier and more efficient**.

```python
import pandas as pd

# Creating a simple DataFrame
data = {
    "Name": ["Arun", "Ravi", "Priya"],
    "Age": [21, 22, 21],
    "Marks": [85, 90, 88]
}

df = pd.DataFrame(data)

print(df)

# Why pd.DataFrame() is used:
# It converts the given structured data into rows and columns.

# Expected Output:
#     Name  Age  Marks
# 0   Arun   21     85
# 1   Ravi   22     90
# 2   Priya  21     88

# AI/ML/Data Analytics Usage:
# DataFrames are commonly used to store, analyze,
# and preprocess datasets before building ML models.
```

# Why Pandas?

## Understand the Concept

Pandas is used because it makes **working with structured data easier, faster, and more organized**.

It provides useful functions for:
- Reading and writing data
- Selecting and filtering data
- Handling missing values
- Cleaning data
- Grouping and analyzing data
- Working with rows and columns

Pandas is especially useful when working with **large datasets** in Data Science and AI/ML.

## Demonstrate the Concept

### Real-World Example

Suppose a company has customer data containing:

Customer Name | Age | Purchase Amount
Arun | 25 | 5000
Ravi | 30 | 7000
Priya | 28 | 6000

Using Pandas, we can easily find customers above a certain age, calculate the average purchase amount, or filter specific records.

Without Pandas, we would have to perform many of these operations manually using Python lists and loops.

### AI/ML Example

In a machine learning project, we may have thousands of rows of customer or sales data.

Pandas helps us to:
- Load the dataset
- Inspect the data
- Clean missing values
- Select required features
- Filter unwanted records
- Prepare data for model training

So, Pandas reduces the effort required for **data preprocessing and analysis**.

```python
import pandas as pd

# Create sample customer data
data = {
    "Name": ["Arun", "Ravi", "Priya"],
    "Age": [25, 30, 28],
    "Purchase": [5000, 7000, 6000]
}

df = pd.DataFrame(data)

# Filter customers whose age is greater than 25
result = df[df["Age"] > 25]

print(result)

# Why filtering is used:
# It helps us select only the rows that satisfy a condition.

# Expected Output:
#     Name  Age  Purchase
# 1   Ravi   30      7000
# 2  Priya   28      6000

# AI/ML/Data Analytics Usage:
# Filtering is commonly used during data preprocessing
# to select relevant records for analysis or model training.
```

# Series

## Understand the Concept

A **Series** is a one-dimensional data structure in Pandas.

It stores a sequence of values along with an **index** for each value.

A Series can contain numbers, strings, or other types of data.

We can create a Series using `pd.Series()`.

In AI/ML and Data Analytics, a Series can represent **one column of a dataset**.

## Demonstrate the Concept

### Real-World Example

Suppose we have the marks of three students:

Student Marks
Arun 85
Ravi 90
Priya 88

This single column of marks can be represented using a Pandas Series.

### AI/ML Example

In a machine learning dataset, a single feature such as **Age**, **Salary**, or **Temperature** can be represented as a Pandas Series.

So, a Series is useful when we are working with **one-dimensional data or a single column**.

```python
import pandas as pd

# Create a Pandas Series
marks = pd.Series([85, 90, 88])

print(marks)

# Why pd.Series() is used:
# It creates a one-dimensional Pandas data structure
# with an index for each value.

# Expected Output:
# 0    85
# 1    90
# 2    88
# dtype: int64

# AI/ML/Data Analytics Usage:
# A Series can represent a single feature or column
# while working with a dataset.
```

# DataFrame

## Understand the Concept

A **DataFrame** is a two-dimensional data structure in Pandas.

It stores data in the form of **rows and columns**, similar to a table in a database or an Excel sheet.

A DataFrame can contain multiple columns with different data types.

In AI/ML, DataFrames are commonly used to **store, analyze, clean, and preprocess datasets**.

## Demonstrate the Concept

### Real-World Example

Suppose we have student information:

Name | Age | Marks
Arun | 21 | 85
Ravi | 22 | 90
Priya | 21 | 88

This complete table can be represented using a Pandas DataFrame.

Each column represents a variable, and each row represents one student's record.

### AI/ML Example

Suppose we have a customer dataset with columns such as:

Customer_ID, Age, Salary, Purchase

We can store the complete dataset in a DataFrame and perform operations such as filtering, cleaning, and analyzing the data before using it for machine learning.

```python
import pandas as pd

# Create a DataFrame using a dictionary
data = {
    "Name": ["Arun", "Ravi", "Priya"],
    "Age": [21, 22, 21],
    "Marks": [85, 90, 88]
}

df = pd.DataFrame(data)

print(df)

# Why pd.DataFrame() is used:
# It converts the dictionary data into a table
# containing rows and columns.

# Expected Output:
#     Name  Age  Marks
# 0   Arun   21     85
# 1   Ravi   22     90
# 2  Priya   21     88

# AI/ML/Data Analytics Usage:
# DataFrames are widely used to store and preprocess
# structured datasets before analysis or model training.
```

# Creating DataFrames

## Understand the Concept

A DataFrame can be created in different ways using Pandas.

We can create a DataFrame from:
- A dictionary
- A list
- A list of dictionaries
- NumPy arrays

The most common method is creating a DataFrame from a **dictionary**, where the keys become column names and the values become column data.

In AI/ML, creating DataFrames is useful for **organizing data before analysis and preprocessing**.

## Demonstrate the Concept

### Real-World Example

Suppose we have employee information such as:

Name | Age | Salary
Arun | 25 | 30000
Ravi | 28 | 40000
Priya | 26 | 35000

We can store this information in a Python dictionary and convert it into a DataFrame.

### AI/ML Example

Suppose we receive customer information from an application as Python data.

We can convert that data into a DataFrame and then perform operations such as filtering, cleaning, and analysis.

So, creating a DataFrame helps us **convert raw structured data into a format that is easy to work with**.

```python
import pandas as pd

# Create data using a dictionary
data = {
    "Name": ["Arun", "Ravi", "Priya"],
    "Age": [25, 28, 26],
    "Salary": [30000, 40000, 35000]
}

# Convert the dictionary into a DataFrame
df = pd.DataFrame(data)

print(df)

# Why pd.DataFrame() is used:
# It converts the dictionary into a tabular structure
# with rows and columns.

# Expected Output:
#     Name  Age  Salary
# 0   Arun   25   30000
# 1   Ravi   28   40000
# 2  Priya   26   35000

# AI/ML/Data Analytics Usage:
# DataFrames are used to organize raw data
# before cleaning, analysis, and model training.
```

# Data Types

## Understand the Concept

Data types tell us **what type of data is stored in each column** of a Pandas DataFrame.

Common Pandas data types include:
- `int64` – integer values
- `float64` – decimal values
- `object` – usually text/string values
- `bool` – True or False values

Knowing the data types is important because different operations require different types of data.

In AI/ML, checking data types helps us **understand the dataset and prepare the features correctly** before model training.

## Demonstrate the Concept

### Real-World Example

Consider employee data:

Name | Age | Salary | Experienced
Arun | 25 | 30000.5 | True
Ravi | 28 | 40000.0 | False

Here:
- Name is text
- Age is an integer
- Salary is a decimal value
- Experienced is a Boolean value

Pandas stores each column according to its data type.

### AI/ML Example

Suppose a machine learning dataset contains Age, Salary, and Customer_Name.

Before training a model, we need to know which columns are numerical and which are text because machine learning algorithms generally require numerical input.

Therefore, checking data types is an important part of data preprocessing.

```python
import pandas as pd

# Create a DataFrame with different types of data
data = {
    "Name": ["Arun", "Ravi", "Priya"],
    "Age": [25, 28, 26],
    "Salary": [30000.5, 40000.0, 35000.75],
    "Experienced": [True, False, True]
}

df = pd.DataFrame(data)

# Check the data types of each column
print(df.dtypes)

# Why dtypes is used:
# It shows the data type of each column in the DataFrame.

# Expected Output:
# Name             object
# Age               int64
# Salary          float64
# Experienced        bool
# dtype: object

# AI/ML/Data Analytics Usage:
# Checking data types helps us identify numerical,
# categorical, and Boolean features during preprocessing.
```

# Index

## Understand the Concept

An **Index** in Pandas is used to identify and access the rows of a Series or DataFrame.

By default, Pandas automatically assigns numbers starting from `0` as the index.

We can also create a **custom index** using meaningful labels.

Indexing helps us easily identify, select, and access specific rows of data.

In AI/ML and Data Analytics, indexes are useful for **organizing and accessing records efficiently**.

## Demonstrate the Concept

### Real-World Example

Suppose we have student marks:

Student | Marks
Arun | 85
Ravi | 90
Priya | 88

Pandas normally assigns indexes as 0, 1, and 2.

We can also use student names as the index:

Arun → 85
Ravi → 90
Priya → 88

This makes it easier to identify a particular student's record.

### AI/ML Example

In a dataset, an index can be used to identify individual records or maintain a unique identifier while performing data analysis and preprocessing.

```python
import pandas as pd

# Create a DataFrame
data = {
    "Marks": [85, 90, 88]
}

df = pd.DataFrame(data, index=["Arun", "Ravi", "Priya"])

print(df)

# Access the marks of Ravi using the index
print("\nRavi's Marks:", df.loc["Ravi", "Marks"])

# Why index is used:
# It provides a label to identify each row
# and helps us access specific records easily.

# Expected Output:
#        Marks
# Arun      85
# Ravi      90
# Priya     88
#
# Ravi's Marks: 90

# AI/ML/Data Analytics Usage:
# Indexes help identify and access specific records
# while working with datasets.
```

# Columns

## Understand the Concept

Columns represent the **individual variables or features** in a Pandas DataFrame.

Each column has a column name and contains values of a particular type.

We can use the `columns` attribute to view all column names in a DataFrame.

We can also select a specific column using its name.

In AI/ML, columns usually represent **features or attributes** that are used for data analysis and model training.

## Demonstrate the Concept

### Real-World Example

Suppose we have employee data:

Name | Age | Salary
Arun | 25 | 30000
Ravi | 28 | 40000
Priya | 26 | 35000

Here, `Name`, `Age`, and `Salary` are the columns.

We can select only the `Salary` column when we want to analyze employee salaries.

### AI/ML Example

Suppose we have a customer dataset containing:

Age | Income | Purchase_Amount

Here, `Age`, `Income`, and `Purchase_Amount` are columns that can represent features or target-related information in a machine learning problem.

Therefore, columns help us **organize and select the required data for analysis**.

```python
import pandas as pd

# Create a DataFrame
data = {
    "Name": ["Arun", "Ravi", "Priya"],
    "Age": [25, 28, 26],
    "Salary": [30000, 40000, 35000]
}

df = pd.DataFrame(data)

# Display all column names
print("Columns:", df.columns)

# Select the Salary column
print("\nSalary Column:")
print(df["Salary"])

# Why columns is used:
# The columns attribute shows all column names.
# Selecting a column helps us work with specific data.

# Expected Output:
# Columns: Index(['Name', 'Age', 'Salary'], dtype='object')
#
# Salary Column:
# 0    30000
# 1    40000
# 2    35000

# AI/ML/Data Analytics Usage:
# Columns represent features or variables that can be
# selected and analyzed during data preprocessing.
```

# Reading Data

## Understand the Concept

Pandas provides functions to **read data from different file formats** and load it into a DataFrame.

Some commonly used functions are:
- `pd.read_csv()` – reads CSV files
- `pd.read_excel()` – reads Excel files
- `pd.read_json()` – reads JSON files

Reading data is usually the first step before **cleaning, analyzing, and preprocessing** a dataset.

In AI/ML, datasets are often stored in files, so Pandas helps us load them into a DataFrame for further processing.

## Demonstrate the Concept

### Real-World Example

Suppose a company stores customer information in a CSV file:

Customer | Age | Purchase
Arun | 25 | 5000
Ravi | 30 | 7000
Priya | 28 | 6000

Using `pd.read_csv()`, we can load this file into a Pandas DataFrame.

### AI/ML Example

Suppose we receive a dataset containing customer information for a machine learning project.

We can use Pandas to read the dataset into a DataFrame and then:
- Inspect the data
- Find missing values
- Clean the data
- Select features
- Prepare the data for model training

So, reading data is an important first step in the data preprocessing process.

```python
import pandas as pd

# Create a sample DataFrame
data = {
    "Name": ["Arun", "Ravi", "Priya"],
    "Age": [25, 30, 28],
    "Purchase": [5000, 7000, 6000]
}

df = pd.DataFrame(data)

# Save the DataFrame as a CSV file
df.to_csv("customers.csv", index=False)

# Read the CSV file using Pandas
read_data = pd.read_csv("customers.csv")

print(read_data)

# Why pd.read_csv() is used:
# It reads data from a CSV file and converts it
# into a Pandas DataFrame.

# Expected Output:
#     Name  Age  Purchase
# 0   Arun   25      5000
# 1   Ravi   30      7000
# 2  Priya   28      6000

# AI/ML/Data Analytics Usage:
# Reading data allows us to load datasets into Pandas
# for cleaning, analysis, and machine learning preprocessing.
```

# Writing Data

## Understand the Concept

Writing data means **saving a Pandas DataFrame into a file**.

Pandas provides different functions to write data into different file formats:
- `to_csv()` – saves data as a CSV file
- `to_excel()` – saves data as an Excel file
- `to_json()` – saves data as a JSON file

Writing data is useful when we want to **store processed or analyzed data for future use**.

In AI/ML, we can save cleaned and processed datasets so they can be reused during model development and analysis.

## Demonstrate the Concept

### Real-World Example

Suppose a company has cleaned its customer data by removing missing or incorrect values.

After cleaning, we can save the DataFrame into a CSV file so that it can be used later.

### AI/ML Example

Suppose we have prepared a dataset for machine learning by:
- Removing missing values
- Selecting important features
- Cleaning incorrect data

We can save this processed DataFrame into a file and use it later for model training.

So, writing data helps us **store and reuse processed information**.

```python
import pandas as pd

# Create a DataFrame
data = {
    "Name": ["Arun", "Ravi", "Priya"],
    "Age": [25, 30, 28],
    "Purchase": [5000, 7000, 6000]
}

df = pd.DataFrame(data)

# Write the DataFrame to a CSV file
df.to_csv("customers_output.csv", index=False)

print("Data successfully written to CSV file.")

# Why to_csv() is used:
# It saves the DataFrame into a CSV file.
# index=False prevents Pandas from saving the DataFrame index
# as an additional column.

# Expected Output:
# Data successfully written to CSV file.

# AI/ML/Data Analytics Usage:
# Processed datasets can be saved and reused later
# for analysis or machine learning model training.
```
