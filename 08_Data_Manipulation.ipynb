# Rename Columns

## Understand the Concept

In Pandas, we use the `rename()` method to change the names of columns in a DataFrame.

We can rename one column or multiple columns without changing the actual data.

### Why do we use it?

- To make column names clear and meaningful.
- To correct spelling mistakes.
- To follow a consistent naming convention.
- To make data easier to understand and work with.

### Syntax

```python
df.rename(columns={"old_name": "new_name"}, inplace=True)

```markdown
## Demonstrate the Concept

### Real-World Example

Imagine a customer dataset with columns:

- `Cust_Name`
- `Cust_Age`
- `Purchase_Amount`

We can rename them to:

- `Customer_Name`
- `Customer_Age`
- `Purchase_Amount`

This makes the column names more readable and consistent.

### AI/ML Example

Suppose a machine learning dataset contains a column named `cust_age`.

Before preprocessing, we can rename it to `customer_age`.

This makes the feature name clearer and easier to understand when performing analysis or building a model.

````python
import pandas as pd

# Create a sample customer DataFrame
df = pd.DataFrame({
    "Cust_Name": ["Arun", "Priya", "Kumar"],
    "Cust_Age": [25, 30, 28],
    "Purchase_Amount": [5000, 7500, 6200]
})

# Display the original column names
print("Before Renaming:")
print(df.columns)

# Rename the columns using rename()
# rename() is used because it allows us to change specific column names.
# inplace=True applies the changes directly to the DataFrame.
df.rename(columns={
    "Cust_Name": "Customer_Name",
    "Cust_Age": "Customer_Age"
}, inplace=True)

# Display the updated DataFrame
print("\nAfter Renaming:")
print(df)

# Expected output:
# The columns Cust_Name and Cust_Age are changed to
# Customer_Name and Customer_Age.

# AI/ML / Data Analytics Use:
# Standardized and meaningful column names make data preprocessing,
# analysis, and machine learning feature selection easier.
````

# Drop Rows

## Understand the Concept

In Pandas, `drop()` is used to remove rows from a DataFrame.

We can remove a single row or multiple rows by specifying their index values.

Dropping rows is useful when the data contains unwanted, incorrect, or irrelevant records.

### Why do we use it?

- To remove unwanted rows.
- To remove incorrect or irrelevant data.
- To clean the dataset before analysis.
- To remove specific records from a DataFrame.

### AI/ML Use

In AI/ML projects, unwanted or irrelevant records can affect data analysis and model performance.

We can use `drop()` to remove such rows during data cleaning and preprocessing.

## Demonstrate the Concept

### Real-World Example

Suppose we have a customer dataset containing three customers.

If one customer's record is incorrect or not required, we can remove that particular row using its index.

For example, if the row with index `1` contains incorrect information, we can drop that row.

### AI/ML Example

Suppose a machine learning dataset contains an invalid customer record.

We can remove that row before training the model so that incorrect data does not affect the analysis or model performance.

````python
import pandas as pd

# Create a sample DataFrame
df = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar", "Divya"],
    "Age": [25, 30, 28, 26],
    "Score": [85, 90, 78, 88]
})

# Display the original DataFrame
print("Original DataFrame:")
print(df)

# Remove the row with index 1
# drop() is used to remove unwanted rows from the DataFrame.
df = df.drop(index=1)

# Display the DataFrame after removing the row
print("\nDataFrame after dropping row:")
print(df)

# Expected output:
# The row with index 1 (Priya) will be removed.

# AI/ML / Data Analytics Use:
# Dropping unwanted or incorrect records is an important
# data-cleaning step before analysis and machine learning.
````

# Drop Columns

## Understand the Concept

In Pandas, `drop()` is used to remove one or more columns from a DataFrame.

We can remove columns by specifying their column names.

Dropping columns is useful when the dataset contains unnecessary, irrelevant, or unwanted features.

### Why do we use it?

- To remove unnecessary columns.
- To remove irrelevant features.
- To reduce the size of the dataset.
- To simplify data analysis.
- To remove columns that are not required for machine learning.

### AI/ML Use

In AI/ML projects, some columns may not provide useful information for the model.

We can remove such columns during data preprocessing to reduce unnecessary data and improve the efficiency of the model.

## Demonstrate the Concept

### Real-World Example

Suppose we have a customer dataset containing:

- `Name`
- `Age`
- `City`
- `Purchase_Amount`

If the customer's name is not required for our analysis, we can remove the `Name` column.

### AI/ML Example

Suppose we are building a machine learning model to predict customer purchase amount.

A column such as `Customer_ID` may only be an identifier and may not provide useful information for prediction.

We can remove the `Customer_ID` column before training the model.

````python
import pandas as pd

# Create a sample DataFrame
df = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar", "Divya"],
    "Age": [25, 30, 28, 26],
    "City": ["Chennai", "Coimbatore", "Salem", "Erode"],
    "Purchase_Amount": [5000, 7500, 6200, 6800]
})

# Display the original DataFrame
print("Original DataFrame:")
print(df)

# Remove the Name column
# drop() is used with axis=1 to remove a column.
df = df.drop("Name", axis=1)

# Display the DataFrame after dropping the column
print("\nDataFrame after dropping Name column:")
print(df)

# Expected output:
# The Name column will be removed from the DataFrame.

# AI/ML / Data Analytics Use:
# Removing unnecessary columns helps reduce irrelevant features
# and makes the dataset easier to analyze and preprocess.
````

# Insert Columns

## Understand the Concept

In Pandas, we can insert a new column into a DataFrame using the `insert()` method.

The `insert()` method allows us to specify the position where the new column should be added.

### Why do we use it?

- To add a new column at a specific position.
- To organize columns in a required order.
- To add a new feature to a dataset.
- To prepare data for analysis or machine learning.

### AI/ML Use

In AI/ML projects, we may create new features from existing data.

For example, we can calculate a customer's age group or total purchase and insert it into a specific position in the DataFrame.

## Demonstrate the Concept

### Real-World Example

Suppose we have a customer dataset containing `Name`, `Age`, and `Purchase_Amount`.

We want to add a new column called `City` between `Name` and `Age`.

Using `insert()`, we can specify the exact position where the new column should be added.

### AI/ML Example

Suppose we have a dataset containing a customer's `Age`.

We can create a new feature called `Age_Group` based on the customer's age and insert it into the DataFrame.

This new feature can be useful during data analysis and machine learning preprocessing.

````python
import pandas as pd

# Create a sample DataFrame
df = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar"],
    "Age": [25, 30, 28],
    "Purchase_Amount": [5000, 7500, 6200]
})

# Display the original DataFrame
print("Original DataFrame:")
print(df)

# Insert a new City column at position 1
# insert() is used when we want to add a column at a specific position.
df.insert(1, "City", ["Chennai", "Coimbatore", "Salem"])

# Display the DataFrame after inserting the column
print("\nDataFrame after inserting City column:")
print(df)

# Expected output:
# The City column will be inserted between Name and Age.

# AI/ML / Data Analytics Use:
# insert() can be used to add newly created features
# at a specific position in a dataset during preprocessing.
````

# assign()

## Understand the Concept

In Pandas, `assign()` is used to create new columns or modify existing columns in a DataFrame.

It returns a new DataFrame with the added or modified columns.

We can use `assign()` to create columns based on existing columns.

### Why do we use it?

- To create new columns easily.
- To modify existing columns.
- To create calculated columns.
- To perform data transformation.
- To make data preprocessing more convenient.

### AI/ML Use

In AI/ML projects, `assign()` can be used to create new features from existing data.

For example, we can calculate a customer's total amount or create a new feature based on existing numerical columns.

## Demonstrate the Concept

### Real-World Example

Suppose we have a customer dataset containing `Price` and `Quantity`.

We can use `assign()` to create a new column called `Total_Amount`.

The `Total_Amount` can be calculated by multiplying `Price` by `Quantity`.

### AI/ML Example

In a machine learning dataset, we can create new features from existing columns.

For example, `Total_Amount` can be created from `Price` and `Quantity`.

This type of feature creation is called feature engineering and can help provide useful information to a machine learning model.

````python
import pandas as pd

# Create a sample DataFrame
df = pd.DataFrame({
    "Product": ["Laptop", "Phone", "Tablet"],
    "Price": [50000, 30000, 20000],
    "Quantity": [2, 3, 4]
})

# Display the original DataFrame
print("Original DataFrame:")
print(df)

# Create a new Total_Amount column using assign()
# assign() is useful for creating new columns from existing columns.
new_df = df.assign(Total_Amount=df["Price"] * df["Quantity"])

# Display the updated DataFrame
print("\nDataFrame after using assign():")
print(new_df)

# Expected output:
# A new Total_Amount column will be created.
# For example, Laptop: 50000 * 2 = 100000.

# AI/ML / Data Analytics Use:
# assign() can be used for feature engineering by creating
# new useful features from existing columns.
````

# apply()

## Understand the Concept

In Pandas, `apply()` is used to apply a function to each value, row, or column of a DataFrame or Series.

It is useful when we want to perform the same operation on multiple values.

### Why do we use it?

- To apply a function to multiple values.
- To transform data.
- To perform calculations on columns.
- To create new values based on existing data.
- To simplify data preprocessing.

### AI/ML Use

In AI/ML projects, `apply()` can be used for data transformation and feature engineering.

For example, we can use `apply()` to convert numerical scores into categories such as `Pass` or `Fail`.

## Demonstrate the Concept

### Real-World Example

Suppose we have a student dataset containing their marks.

We can use `apply()` to check each student's mark and classify them as:

- `Pass` if the mark is 50 or above.
- `Fail` if the mark is below 50.

The same function is applied to every value in the marks column.

### AI/ML Example

In a machine learning dataset, we may need to transform numerical values into meaningful categories.

For example, customer spending amounts can be converted into categories such as `High`, `Medium`, or `Low`.

This type of transformation can be useful during data preprocessing and feature engineering.

````python
import pandas as pd

# Create a sample DataFrame
df = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar", "Divya"],
    "Marks": [85, 45, 72, 38]
})

# Display the original DataFrame
print("Original DataFrame:")
print(df)

# Define a function to classify marks
def check_result(mark):
    if mark >= 50:
        return "Pass"
    else:
        return "Fail"

# Apply the function to every value in the Marks column
# apply() is used because we want the same function
# to be applied to each mark.
df["Result"] = df["Marks"].apply(check_result)

# Display the updated DataFrame
print("\nDataFrame after using apply():")
print(df)

# Expected output:
# Marks 85 -> Pass
# Marks 45 -> Fail
# Marks 72 -> Pass
# Marks 38 -> Fail

# AI/ML / Data Analytics Use:
# apply() is useful for transforming data and creating
# new features during data preprocessing and feature engineering.
````

# map()

## Understand the Concept

In Pandas, `map()` is used to apply a function or mapping to each value in a Series.

It is mainly used to transform or replace individual values in a column.

### Why do we use it?

- To transform values in a column.
- To replace values with meaningful labels.
- To perform value-based mapping.
- To clean and standardize categorical data.

### AI/ML Use

In AI/ML projects, `map()` can be used to convert categorical values into meaningful or numerical values.

For example, `Yes` and `No` can be mapped to `1` and `0` before using the data for machine learning.

## Demonstrate the Concept

### Real-World Example

Suppose we have a customer dataset with a column called `Membership`.

The values are:

- `Yes`
- `No`

We can use `map()` to convert them into:

- `Yes` → `1`
- `No` → `0`

This makes the data easier to process.

### AI/ML Example

Machine learning models generally work with numerical data.

Therefore, categorical values such as `Yes` and `No` can be converted into `1` and `0` using `map()` during data preprocessing.

````python
import pandas as pd

# Create a sample DataFrame
df = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar", "Divya"],
    "Membership": ["Yes", "No", "Yes", "No"]
})

# Display the original DataFrame
print("Original DataFrame:")
print(df)

# Map Yes to 1 and No to 0
# map() is used to replace each value based on a mapping dictionary.
df["Membership"] = df["Membership"].map({
    "Yes": 1,
    "No": 0
})

# Display the updated DataFrame
print("\nDataFrame after using map():")
print(df)

# Expected output:
# Yes will be converted to 1.
# No will be converted to 0.

# AI/ML / Data Analytics Use:
# map() can be used to convert categorical values into
# numerical values during machine learning preprocessing.
````

# replace()

## Understand the Concept

In Pandas, `replace()` is used to replace specific values in a DataFrame or Series with new values.

It can be used to replace one value or multiple values.

### Why do we use it?

- To correct incorrect values.
- To replace old values with new values.
- To clean and standardize data.
- To handle inconsistent values in a dataset.

### AI/ML Use

In AI/ML projects, datasets may contain inconsistent or incorrect values.

We can use `replace()` to clean and standardize these values before performing data analysis or training a machine learning model.

## Demonstrate the Concept

### Real-World Example

Suppose a customer dataset contains city names such as:

- `Chennai`
- `Chennai`
- `Madras`

Here, `Madras` refers to the same city as `Chennai`.

We can use `replace()` to change `Madras` to `Chennai` and make the data consistent.

### AI/ML Example

In an AI/ML dataset, categorical values may be written in different ways.

For example, `Male`, `M`, and `male` may represent the same category.

Using `replace()`, we can standardize these values before model training.

````python
import pandas as pd

# Create a sample DataFrame
df = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar", "Divya"],
    "City": ["Chennai", "Madras", "Salem", "Madras"]
})

# Display the original DataFrame
print("Original DataFrame:")
print(df)

# Replace Madras with Chennai
# replace() is used to change specific values in the DataFrame.
df["City"] = df["City"].replace("Madras", "Chennai")

# Display the updated DataFrame
print("\nDataFrame after using replace():")
print(df)

# Expected output:
# All occurrences of Madras will be changed to Chennai.

# AI/ML / Data Analytics Use:
# replace() helps standardize inconsistent categorical values
# during data cleaning and preprocessing.
````

# Duplicate Handling

## Understand the Concept

In Pandas, duplicate handling is used to identify and remove duplicate rows from a DataFrame.

We use `duplicated()` to find duplicate rows and `drop_duplicates()` to remove them.

Duplicate records can occur when the same data is entered or collected more than once.

### Why do we use it?

- To identify duplicate records.
- To remove repeated rows.
- To improve data quality.
- To avoid incorrect analysis caused by repeated data.

### AI/ML Use

In AI/ML projects, duplicate records can affect statistical analysis and machine learning model performance.

Removing duplicate data helps ensure that each record is represented correctly in the dataset.

## Demonstrate the Concept

### Real-World Example

Suppose a customer dataset contains the same customer's information more than once.

For example, if Arun's record appears twice, it is a duplicate record.

We can use `duplicated()` to identify the duplicate and `drop_duplicates()` to remove it.

### AI/ML Example

Suppose a machine learning dataset contains duplicate training records.

The model may see the same information multiple times, which can affect the learning process.

Removing duplicate records helps create a cleaner dataset for model training.

````python
import pandas as pd

# Create a sample DataFrame with a duplicate row
df = pd.DataFrame({
    "Name": ["Arun", "Priya", "Arun", "Kumar"],
    "Age": [25, 30, 25, 28],
    "Score": [85, 90, 85, 78]
})

# Display the original DataFrame
print("Original DataFrame:")
print(df)

# Check for duplicate rows
# duplicated() returns True for rows that are duplicates.
print("\nDuplicate Rows:")
print(df.duplicated())

# Remove duplicate rows
# drop_duplicates() removes repeated rows from the DataFrame.
df = df.drop_duplicates()

# Display the DataFrame after removing duplicates
print("\nDataFrame after removing duplicates:")
print(df)

# Expected output:
# The second occurrence of Arun's record will be removed.

# AI/ML / Data Analytics Use:
# Removing duplicate records improves data quality and helps
# prevent repeated observations from affecting analysis or model training.
````
