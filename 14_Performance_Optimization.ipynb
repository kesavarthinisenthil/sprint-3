# Vectorization

## Understand the Concept

In Pandas, vectorization means performing an operation on an entire Series or DataFrame at once instead of using a Python loop to process each value individually.

Vectorized operations are usually faster and more efficient because Pandas and NumPy perform the operation internally on the whole data.

### Why do we use it?

- To process many values at once.
- To make code shorter and simpler.
- To improve performance.
- To handle large datasets efficiently.

### AI/ML Use

In AI/ML projects, datasets often contain thousands or millions of rows.

Vectorization can be used for operations such as scaling values, creating new features, or performing mathematical calculations efficiently.

## Demonstrate the Concept

### Real-World Example

Suppose a company has sales values for thousands of products and wants to increase every price by 10%.

Instead of looping through every row, we can multiply the entire Pandas column by `1.10` at once.

### AI/ML Example

Suppose an ML dataset contains a `Salary` column and we want to convert salaries from rupees to thousands.

We can divide the entire column by `1000` using a vectorized operation, without writing a loop.

```python
import pandas as pd

# Create a DataFrame containing product prices
data = pd.DataFrame({
    "Product": ["Laptop", "Phone", "Tablet"],
    "Price": [50000, 30000, 20000]
})

# Display the original data
print("Original Data:")
print(data)

# Increase all prices by 10% using a vectorized operation
# Multiplying the entire column processes all values at once.
data["Updated_Price"] = data["Price"] * 1.10

# Display the updated data
print("\nAfter Vectorized Operation:")
print(data)

# Expected output:
# Laptop -> 55000.0
# Phone -> 33000.0
# Tablet -> 22000.0

# AI/ML / Data Analytics Use:
# Vectorization is useful for fast calculations and
# feature engineering on large datasets.
```

# apply() vs Loops

## Understand the Concept

In Pandas, `apply()` is used to apply a function to each value or row of a Series or DataFrame.

A Python loop can also process values one by one, but `apply()` provides a simpler way to apply a function to Pandas data.

For many common calculations, vectorization is usually faster than both loops and `apply()`.

### Why do we use it?

- To apply a custom function to Pandas data.
- To perform operations that are not easily written as simple vectorized expressions.
- To make data transformation easier.
- To process values or rows using a function.

### AI/ML Use

In AI/ML projects, `apply()` can be used for feature engineering and data transformation when a custom rule or function is required.

## Demonstrate the Concept

### Real-World Example

Suppose a company has employee salaries and wants to classify employees as "High Salary" or "Normal Salary".

We can create a function and use `apply()` to apply that function to every salary value.

A loop could also do this, but `apply()` provides a simple Pandas-based approach.

### AI/ML Example

Suppose an ML dataset contains customer ages.

We may want to create an age group such as "Young" or "Adult" based on a custom condition.

We can use `apply()` with a function to create this new feature.

```python
import pandas as pd

# Create a DataFrame containing employee salaries
data = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar", "Divya"],
    "Salary": [25000, 60000, 35000, 75000]
})

# Display the original data
print("Original Data:")
print(data)

# Create a function to classify salary
def salary_category(salary):
    if salary >= 50000:
        return "High Salary"
    else:
        return "Normal Salary"

# Apply the function to the Salary column
# apply() processes each value using the custom function.
data["Category"] = data["Salary"].apply(salary_category)

# Display the updated DataFrame
print("\nAfter Using apply():")
print(data)

# Expected output:
# Salaries >= 50000 -> High Salary
# Salaries < 50000 -> Normal Salary

# AI/ML / Data Analytics Use:
# apply() can be useful for custom feature creation
# and data transformation when a simple vectorized
# operation is not suitable.
```

# Memory Optimization

## Understand the Concept

Memory optimization means reducing the amount of memory used by a Pandas DataFrame while keeping the required data.

Large datasets can use a lot of memory, especially when they contain many rows and columns.

We can optimize memory by choosing suitable data types for columns and avoiding unnecessary data.

### Why do we use it?

- To reduce memory usage.
- To handle large datasets more efficiently.
- To improve application performance.
- To avoid memory-related problems.

### AI/ML Use

In AI/ML projects, datasets can contain millions of rows.

Memory optimization helps us load, process, and analyze large datasets efficiently, especially when working with limited system memory.

## Demonstrate the Concept

### Real-World Example

Suppose a company has a large customer dataset.

If a column contains only small integer values, storing those values using a large integer data type may use more memory than necessary.

We can choose a suitable data type to reduce memory usage.

### AI/ML Example

Suppose an ML dataset contains customer ages from 18 to 80.

Instead of using a large integer type unnecessarily, we can use a smaller integer type when appropriate.

This can reduce memory usage when working with large datasets.

```python
import pandas as pd

# Create a DataFrame
data = pd.DataFrame({
    "Age": [22, 25, 30, 35, 40],
    "Salary": [25000, 30000, 40000, 50000, 60000]
})

# Check memory usage before optimization
print("Memory Usage Before Optimization:")
print(data.memory_usage(deep=True))

# Convert Age to a smaller integer data type
# int8 is suitable for values within its supported range.
data["Age"] = data["Age"].astype("int8")

# Check memory usage after optimization
print("\nMemory Usage After Optimization:")
print(data.memory_usage(deep=True))

# Display the DataFrame
print("\nData:")
print(data)

# Expected output:
# The Age column uses less memory after changing
# its data type to int8.

# AI/ML / Data Analytics Use:
# Choosing suitable data types can reduce memory usage
# when processing large datasets.
```

# Category dtype

## Understand the Concept

In Pandas, `category` is a data type used for columns that contain a limited number of repeated values.

For example, columns such as Gender, City, Department, or Product Type may contain the same values many times.

Using `category` can reduce memory usage and make some operations more efficient.

### Why do we use it?

- To reduce memory usage.
- To efficiently store repeated values.
- To improve performance for some operations.
- To represent categorical data clearly.

### AI/ML Use

In AI/ML projects, datasets often contain categorical features such as City, Gender, Department, or Customer Type.

Using the `category` dtype can help reduce memory usage when these columns contain many repeated values.

## Demonstrate the Concept

### Real-World Example

Suppose a company has thousands of employee records.

The Department column may contain only a few values such as "HR", "Sales", "IT", and "Finance".

Instead of storing these repeated text values normally, we can use the `category` dtype.

### AI/ML Example

Suppose an ML dataset contains customer information and the `City` column contains only a few repeated city names.

We can convert the City column to `category` to reduce memory usage and handle the categorical data efficiently.

```python
import pandas as pd

# Create a DataFrame with repeated categorical values
data = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar", "Divya", "Ravi"],
    "Department": ["IT", "HR", "IT", "Sales", "IT"]
})

# Display the original data types
print("Data Types Before Optimization:")
print(data.dtypes)

# Convert Department to category dtype
# category is useful when a column contains repeated values.
data["Department"] = data["Department"].astype("category")

# Display the updated data types
print("\nData Types After Optimization:")
print(data.dtypes)

# Display the data
print("\nData:")
print(data)

# Expected output:
# Department will have the dtype "category".

# AI/ML / Data Analytics Use:
# category is useful for storing repeated categorical
# values efficiently in large datasets.
```

# Efficient Pandas Practices

## Understand the Concept

Efficient Pandas practices mean writing Pandas code in a way that uses less memory, runs faster, and is easier to maintain.

Instead of using unnecessary loops or loading unwanted data, we should use Pandas functions and select only the data we need.

### Why do we use it?

- To improve code performance.
- To reduce memory usage.
- To process large datasets efficiently.
- To make code simpler and easier to maintain.

### AI/ML Use

In AI/ML projects, datasets can contain thousands or millions of records.

Efficient Pandas practices help us clean, transform, and prepare large datasets faster before training machine learning models.

## Demonstrate the Concept

### Real-World Example

Suppose a company has a large sales dataset with many columns, but we only need the Product, Sales, and Region columns for analysis.

Instead of loading or processing unnecessary columns, we can work only with the required columns.

### AI/ML Example

Suppose an ML dataset contains many features, but our model only needs a few relevant features.

Selecting only the required columns can reduce memory usage and make data preprocessing more efficient.

```python
import pandas as pd

# Create a sample sales DataFrame
data = pd.DataFrame({
    "Product": ["Laptop", "Phone", "Tablet"],
    "Sales": [50000, 30000, 20000],
    "Region": ["Chennai", "Salem", "Coimbatore"],
    "Unnecessary_Data": [100, 200, 300]
})

# Display the original DataFrame
print("Original Data:")
print(data)

# Select only the required columns
# Selecting required columns avoids unnecessary data processing.
required_data = data[["Product", "Sales", "Region"]]

# Display the optimized DataFrame
print("\nRequired Data:")
print(required_data)

# Expected output:
# The Unnecessary_Data column is removed from the data
# being processed.

# AI/ML / Data Analytics Use:
# Selecting only required columns can reduce memory usage
# and improve efficiency during data preprocessing.
```
