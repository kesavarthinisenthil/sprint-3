# loc

## Understand the Concept

The `loc` indexer is used to select rows and columns from a Pandas DataFrame using labels.

We can use `loc` to:
- Select a specific row using its index label
- Select specific columns by their names
- Select rows and columns together
- Filter data using conditions

The general syntax is:

`df.loc[row_label, column_label]`

`loc` works mainly with labels rather than numerical positions.

## Demonstrate the Concept

### Real-World Example

Suppose we have customer data with columns such as `Name`, `Age`, and `Purchase`.

If we want to select the customer with index `2`, we can use `loc`.

We can also select only the `Name` and `Purchase` columns for a particular customer.

### AI/ML Example

In AI/ML data preprocessing, `loc` can be used to select specific features or rows based on conditions.

For example, we can select customers whose purchase amount is greater than 500 and use those records for further analysis.

```python
import pandas as pd

# Create a sample customer DataFrame.
data = {
    "Name": ["Arun", "Priya", "Kumar", "Divya", "Ravi"],
    "Age": [25, 28, 30, 24, 32],
    "Purchase": [500, 750, 1000, 450, 1200]
}

df = pd.DataFrame(data)

# Select the row with index label 2.
# loc selects data using row/column labels.
print("Row with index 2:")
print(df.loc[2])

# Select specific rows and columns.
# Here, we select rows 1 and 3 and only the Name and Purchase columns.
print("\nSelected Rows and Columns:")
print(df.loc[[1, 3], ["Name", "Purchase"]])

# Select rows based on a condition.
# loc can combine label-based selection with a Boolean condition.
print("\nCustomers with Purchase greater than 500:")
print(df.loc[df["Purchase"] > 500, ["Name", "Purchase"]])

# Expected Output:
# Row with index 2:
# Name          Kumar
# Age              30
# Purchase        1000
#
# Selected Rows and Columns:
#     Name  Purchase
# 1  Priya       750
# 3  Divya       450

# AI/ML / Data Analytics Use:
# loc is useful for selecting specific records and features
# during data cleaning, filtering, and Machine Learning preprocessing.
```

# iloc

## Understand the Concept

The `iloc` indexer is used to select rows and columns from a Pandas DataFrame based on their numerical positions.

`iloc` uses integer positions starting from `0`.

The general syntax is:

`df.iloc[row_position, column_position]`

For example:
- `iloc[0]` selects the first row.
- `iloc[:, 0]` selects the first column.
- `iloc[0:3, 0:2]` selects the first 3 rows and first 2 columns.

Unlike `loc`, `iloc` works with positions rather than labels.

## Demonstrate the Concept

### Real-World Example

Suppose we have a customer DataFrame with several rows and columns.

If we want to select the first three customers and the first two columns, we can use `iloc` based on their numerical positions.

### AI/ML Example

During data preprocessing, `iloc` can be useful when we need to select data based on its position.

For example, we can select the first few rows for testing or select a group of feature columns based on their positions.

```python
import pandas as pd

# Create a sample customer DataFrame.
data = {
    "Name": ["Arun", "Priya", "Kumar", "Divya", "Ravi"],
    "Age": [25, 28, 30, 24, 32],
    "Purchase": [500, 750, 1000, 450, 1200]
}

df = pd.DataFrame(data)

# Select the first row.
# iloc[0] selects the row at position 0.
print("First Row:")
print(df.iloc[0])

# Select the first three rows.
# iloc[0:3] selects rows at positions 0, 1, and 2.
print("\nFirst Three Rows:")
print(df.iloc[0:3])

# Select the first three rows and first two columns.
# iloc[0:3, 0:2] selects rows 0-2 and columns 0-1.
print("\nSelected Rows and Columns:")
print(df.iloc[0:3, 0:2])

# Expected Output:
# First Row:
# Name          Arun
# Age             25
# Purchase       500
#
# First Three Rows:
#     Name  Age  Purchase
# 0   Arun   25       500
# 1  Priya   28       750
# 2  Kumar   30      1000
#
# Selected Rows and Columns:
#     Name  Age
# 0   Arun   25
# 1  Priya   28
# 2  Kumar   30

# AI/ML / Data Analytics Use:
# iloc is useful when selecting rows or columns based on
# their numerical position during data preprocessing and analysis.
```

# Boolean Indexing

## Understand the Concept

Boolean indexing is a technique used to select rows from a Pandas DataFrame based on a condition.

The condition produces Boolean values:

- `True` – the row satisfies the condition
- `False` – the row does not satisfy the condition

Pandas returns only the rows where the condition is `True`.

For example:

`df[df["Age"] > 25]`

This selects customers whose age is greater than 25.

## Demonstrate the Concept

### Real-World Example

Suppose a company has customer data and wants to find customers whose purchase amount is greater than 500.

We can create a condition on the `Purchase` column and use Boolean indexing to select only those customers.

### AI/ML Example

Boolean indexing is commonly used during data preprocessing.

For example, we can select customers with age greater than 25, remove records with invalid values, or select data that satisfies a particular condition before training a Machine Learning model.

```python
import pandas as pd

# Create a sample customer DataFrame.
data = {
    "Name": ["Arun", "Priya", "Kumar", "Divya", "Ravi"],
    "Age": [25, 28, 30, 24, 32],
    "Purchase": [500, 750, 1000, 450, 1200]
}

df = pd.DataFrame(data)

# Create a Boolean condition.
# This checks whether the Purchase value is greater than 500.
condition = df["Purchase"] > 500

# Display the Boolean values.
print("Boolean Condition:")
print(condition)

# Use Boolean indexing to select only matching rows.
print("\nCustomers with Purchase greater than 500:")
print(df[condition])

# We can also write the condition directly.
print("\nCustomers with Age greater than 25:")
print(df[df["Age"] > 25])

# Expected Output:
# Customers with Purchase greater than 500:
#     Name  Age  Purchase
# 1  Priya   28       750
# 2  Kumar   30      1000
# 4   Ravi   32      1200

# AI/ML / Data Analytics Use:
# Boolean indexing is useful for filtering datasets,
# removing invalid records, selecting relevant samples,
# and preparing data for Machine Learning.
```

# Filtering Rows

## Understand the Concept

Filtering rows means selecting only the rows that satisfy a specific condition.

In Pandas, we can filter rows using conditions on one or more columns.

For example:

`df[df["Age"] > 25]`

This returns only the rows where the `Age` is greater than 25.

We can also combine multiple conditions using:
- `&` – AND
- `|` – OR
- `~` – NOT

## Demonstrate the Concept

### Real-World Example

Suppose a company wants to find customers who are above 25 years old and have made purchases greater than 500.

We can combine two conditions to filter only those customers.

### AI/ML Example

Filtering is commonly used during data preprocessing.

For example, we can select customers from a particular age group, remove invalid records, or select samples that meet specific business conditions before performing Machine Learning.

```python
import pandas as pd

# Create a sample customer DataFrame.
data = {
    "Name": ["Arun", "Priya", "Kumar", "Divya", "Ravi", "Anu"],
    "Age": [25, 28, 30, 24, 32, 27],
    "Purchase": [500, 750, 1000, 450, 1200, 800]
}

df = pd.DataFrame(data)

# Filter rows where Age is greater than 25.
print("Age greater than 25:")
print(df[df["Age"] > 25])

# Filter rows using two conditions.
# & means both conditions must be True.
filtered_data = df[(df["Age"] > 25) & (df["Purchase"] > 700)]

print("\nAge greater than 25 AND Purchase greater than 700:")
print(filtered_data)

# Filter rows using OR.
# | means at least one condition must be True.
filtered_data_or = df[(df["Age"] < 25) | (df["Purchase"] > 1000)]

print("\nAge less than 25 OR Purchase greater than 1000:")
print(filtered_data_or)

# Expected Output:
# The output will contain only the rows
# that satisfy the given conditions.

# AI/ML / Data Analytics Use:
# Filtering is useful for data cleaning, feature analysis,
# selecting relevant records, and preparing datasets
# for Machine Learning.
```

# query()

## Understand the Concept

The `query()` method is used to filter rows in a Pandas DataFrame using a query expression.

It provides a clean and readable way to filter data, especially when working with multiple conditions.

For example:

`df.query("Age > 25")`

This selects rows where the `Age` is greater than 25.

For multiple conditions, we can use `and` and `or` inside the query expression.

## Demonstrate the Concept

### Real-World Example

Suppose a company wants to find customers who are above 25 years old and have made purchases greater than 700.

Instead of writing a long Boolean condition, we can use `query()` to write the condition in a simple and readable form.

### AI/ML Example

In AI/ML data preprocessing, `query()` can be used to quickly filter records based on feature values.

For example, we can select customers with a particular age range or purchase amount before performing further analysis.

```python
import pandas as pd

# Create a sample customer DataFrame.
data = {
    "Name": ["Arun", "Priya", "Kumar", "Divya", "Ravi", "Anu"],
    "Age": [25, 28, 30, 24, 32, 27],
    "Purchase": [500, 750, 1000, 450, 1200, 800]
}

df = pd.DataFrame(data)

# Filter customers whose Age is greater than 25.
# query() provides a readable way to filter DataFrame rows.
print("Age greater than 25:")
print(df.query("Age > 25"))

# Filter using multiple conditions.
# 'and' means both conditions must be True.
print("\nAge greater than 25 AND Purchase greater than 700:")
print(df.query("Age > 25 and Purchase > 700"))

# Filter using OR.
# 'or' means at least one condition must be True.
print("\nAge less than 25 OR Purchase greater than 1000:")
print(df.query("Age < 25 or Purchase > 1000"))

# Expected Output:
# The output contains only the rows
# that satisfy each query condition.

# AI/ML / Data Analytics Use:
# query() is useful for readable and efficient data filtering
# during Exploratory Data Analysis and preprocessing.
```

# isin()

## Understand the Concept

The `isin()` method is used to check whether values in a Pandas column belong to a given list of values.

It returns `True` when a value is present in the specified list and `False` when it is not.

For example:

`df["City"].isin(["Chennai", "Salem"])`

checks whether each city is either Chennai or Salem.

It is useful when we want to filter rows based on multiple specific values.

## Demonstrate the Concept

### Real-World Example

Suppose a company wants to analyse customers from only Chennai and Salem.

Instead of writing multiple OR conditions, we can use `isin()` with a list of cities.

### AI/ML Example

In AI/ML preprocessing, `isin()` can be used to select records belonging to specific categories.

For example, we can select customers from selected regions, departments, or product categories before performing further analysis or model preparation.

```python
import pandas as pd

# Create a sample customer DataFrame.
data = {
    "Name": ["Arun", "Priya", "Kumar", "Divya", "Ravi", "Anu"],
    "City": ["Chennai", "Coimbatore", "Chennai", "Salem", "Madurai", "Salem"],
    "Purchase": [500, 750, 1000, 450, 1200, 800]
}

df = pd.DataFrame(data)

# Check whether each city belongs to the selected cities.
# isin() checks values against a list of values.
condition = df["City"].isin(["Chennai", "Salem"])

print("Boolean Condition:")
print(condition)

# Filter rows using the isin() condition.
print("\nCustomers from Chennai or Salem:")
print(df[condition])

# Expected Output:
# Customers from Chennai or Salem:
#     Name      City  Purchase
# 0   Arun   Chennai       500
# 2  Kumar   Chennai      1000
# 3  Divya     Salem       450
# 5    Anu     Salem       800

# AI/ML / Data Analytics Use:
# isin() is useful for filtering selected categories,
# regions, departments, or other categorical values
# during data analysis and preprocessing.
```

# between()

## Understand the Concept

The `between()` method is used to check whether values in a Pandas Series fall within a specified range.

It returns `True` for values within the given range and `False` for values outside the range.

For example:

`df["Age"].between(25, 30)`

checks whether the age is between 25 and 30.

By default, both boundary values are included.

## Demonstrate the Concept

### Real-World Example

Suppose a company wants to find customers whose age is between 25 and 30.

Instead of writing two separate conditions, we can use `between()` to filter the required age range.

### AI/ML Example

In AI/ML data preprocessing, `between()` can be used to select records within a specific numerical range.

For example, we can select customers whose income, age, score, or purchase amount falls within a particular range before further analysis.

```python
import pandas as pd

# Create a sample customer DataFrame.
data = {
    "Name": ["Arun", "Priya", "Kumar", "Divya", "Ravi", "Anu"],
    "Age": [22, 25, 28, 30, 35, 40],
    "Purchase": [500, 750, 1000, 450, 1200, 800]
}

df = pd.DataFrame(data)

# Check which customers have an age between 25 and 30.
# between() checks whether values fall within the given range.
condition = df["Age"].between(25, 30)

print("Boolean Condition:")
print(condition)

# Filter the DataFrame using the condition.
print("\nCustomers with Age between 25 and 30:")
print(df[condition])

# Expected Output:
# Customers with Age between 25 and 30:
#     Name  Age  Purchase
# 1  Priya   25       750
# 2  Kumar   28      1000
# 3  Divya   30       450

# AI/ML / Data Analytics Use:
# between() is useful for filtering numerical features
# within a specific range during data analysis and preprocessing.
```

# Sorting Data

## Understand the Concept

Sorting data means arranging the rows of a Pandas DataFrame based on one or more columns.

Pandas provides the `sort_values()` method for sorting data.

We can sort data:
- In ascending order
- In descending order
- Using one column
- Using multiple columns

For example:

`df.sort_values("Age")`

sorts the DataFrame by age in ascending order.

For descending order, we use:

`df.sort_values("Age", ascending=False)``df.sort_values("Age", ascending=False)`

## Demonstrate the Concept

### Real-World Example

Suppose a company wants to identify customers with the highest purchase amounts.

We can sort the `Purchase` column in descending order so that customers with the highest purchases appear first.

### AI/ML Example

Sorting is useful during Exploratory Data Analysis and data preprocessing.

For example, we can sort records by a feature to identify the highest or lowest values, inspect trends, or prepare data for further analysis.

```python
import pandas as pd

# Create a sample customer DataFrame.
data = {
    "Name": ["Arun", "Priya", "Kumar", "Divya", "Ravi"],
    "Age": [25, 28, 30, 24, 32],
    "Purchase": [500, 750, 1000, 450, 1200]
}

df = pd.DataFrame(data)

# Sort by Age in ascending order.
# sort_values() arranges rows based on the selected column.
print("Sorted by Age - Ascending:")
print(df.sort_values("Age"))

# Sort by Purchase in descending order.
# ascending=False arranges values from highest to lowest.
print("\nSorted by Purchase - Descending:")
print(df.sort_values("Purchase", ascending=False))

# Sort by multiple columns.
# First sort by Age and then by Purchase.
print("\nSorted by Age and Purchase:")
print(df.sort_values(["Age", "Purchase"]))

# Expected Output:
# The rows will be arranged according to the selected
# column or columns and the specified sorting order.

# AI/ML / Data Analytics Use:
# Sorting is useful during EDA to identify highest/lowest
# values, understand patterns, and inspect important records.
```
