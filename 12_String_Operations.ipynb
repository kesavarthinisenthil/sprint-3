# lower()

## Understand the Concept

In Pandas, `lower()` is used to convert all characters in a string to lowercase.

It is useful for making text values consistent when they contain a mixture of uppercase and lowercase letters.

### Why do we use it?

- To convert text into lowercase.
- To make text values consistent.
- To make text comparison easier.
- To clean text data before analysis.

### AI/ML Use

In AI/ML projects, text data may contain the same word in different cases.

For example, `Python`, `PYTHON`, and `python` may represent the same value.

Converting text to lowercase helps standardize the data before text analysis or machine learning.

## Demonstrate the Concept

### Real-World Example

Suppose a company stores customer names or city names in different cases.

For example:

- Chennai
- CHENNAI
- chennai

We can use `lower()` to convert all values into lowercase and make them consistent.

### AI/ML Example

Suppose a dataset contains customer feedback such as:

- "GOOD PRODUCT"
- "Good Product"
- "good product"

Converting the text to lowercase makes these values consistent before performing text analysis or feature engineering.

```python
import pandas as pd

# Create a DataFrame with text values in different cases
data = pd.DataFrame({
    "City": ["Chennai", "COIMBATORE", "Bangalore"]
})

# Display the original data
print("Original Data:")
print(data)

# Convert all city names to lowercase
# str.lower() applies the lower() operation to each string.
data["City_Lower"] = data["City"].str.lower()

# Display the converted data
print("\nAfter using lower():")
print(data)

# Expected output:
# Chennai -> chennai
# COIMBATORE -> coimbatore
# Bangalore -> bangalore

# AI/ML / Data Analytics Use:
# Converting text to lowercase helps standardize text
# before analysis, searching, or machine learning.
```

# upper()

## Understand the Concept

In Pandas, `upper()` is used to convert all characters in a string to uppercase.

It is useful when we want to make text values consistent by converting them into the same uppercase format.

### Why do we use it?

- To convert text into uppercase.
- To make text values consistent.
- To make text comparison easier.
- To standardize text data.

### AI/ML Use

In AI/ML projects, text data may contain values in different cases.

For example, `python`, `Python`, and `PYTHON` may represent the same value.

Converting text to uppercase helps standardize the data before analysis or machine learning.

## Demonstrate the Concept

### Real-World Example

Suppose a company stores department names in different cases.

For example:

- sales
- Sales
- SALES

We can use `upper()` to convert all values into uppercase and make them consistent.

### AI/ML Example

Suppose a customer dataset contains product categories such as `laptop`, `Laptop`, and `LAPTOP`.

Converting these values to uppercase helps create consistent categories before data analysis or machine learning.

```python
import pandas as pd

# Create a DataFrame with text values in different cases
data = pd.DataFrame({
    "Department": ["sales", "Marketing", "finance"]
})

# Display the original data
print("Original Data:")
print(data)

# Convert all department names to uppercase
# str.upper() applies the upper() operation to each string.
data["Department_Upper"] = data["Department"].str.upper()

# Display the converted data
print("\nAfter using upper():")
print(data)

# Expected output:
# sales -> SALES
# Marketing -> MARKETING
# finance -> FINANCE

# AI/ML / Data Analytics Use:
# Converting text to uppercase helps standardize categorical
# and text data before analysis or machine learning.
```

# strip()

## Understand the Concept

In Pandas, `strip()` is used to remove unwanted spaces from the beginning and end of a string.

It is useful for cleaning text data that may contain extra spaces.

### Why do we use it?

- To remove unnecessary spaces.
- To clean text data.
- To make text values consistent.
- To avoid incorrect comparisons caused by extra spaces.

### AI/ML Use

In AI/ML projects, datasets may contain unwanted spaces in names, categories, or other text fields.

Using `strip()` helps clean the data before analysis, filtering, or machine learning.

## Demonstrate the Concept

### Real-World Example

Suppose a customer dataset contains names with extra spaces.

For example:

- `" Arun"`
- `"Priya "`
- `" Kumar "`

These extra spaces can cause problems when comparing or grouping values.

We can use `strip()` to remove the spaces from both sides of the text.

### AI/ML Example

Suppose a dataset contains product categories such as `" Laptop "`, `"Laptop"`, and `" Laptop"`.

Using `strip()` makes these values consistent and helps avoid duplicate categories during data preprocessing.

```python
import pandas as pd

# Create a DataFrame with extra spaces
data = pd.DataFrame({
    "Name": [" Arun ", "Priya  ", "  Kumar"]
})

# Display the original data
print("Original Data:")
print(data)

# Remove spaces from the beginning and end of each name
# str.strip() removes unwanted leading and trailing spaces.
data["Clean_Name"] = data["Name"].str.strip()

# Display the cleaned data
print("\nAfter using strip():")
print(data)

# Expected output:
# " Arun "  -> "Arun"
# "Priya  " -> "Priya"
# "  Kumar" -> "Kumar"

# AI/ML / Data Analytics Use:
# strip() helps clean text data and prevents extra spaces
# from creating inconsistent or duplicate values.
```

# split()

## Understand the Concept

In Pandas, `split()` is used to divide a string into smaller parts based on a separator.

It is useful when a single text value contains multiple pieces of information.

### Why do we use it?

- To divide text into smaller parts.
- To separate words or values.
- To extract useful information from text.
- To prepare text data for analysis.

### AI/ML Use

In AI/ML projects, text data may contain multiple values in one column.

For example, a full name can be separated into first name and last name.

This helps create useful features from text data during preprocessing.

## Demonstrate the Concept

### Real-World Example

Suppose a company stores a customer's full name in one column.

For example:

`"Arun Kumar"`

We can use `split()` to separate the name into:

- Arun
- Kumar

This makes it easier to work with each part separately.

### AI/ML Example

Suppose a dataset contains product categories such as:

`"Electronics-Mobile"`

We can use `split()` to separate the main category and subcategory.

These separate values can then be used as features during data analysis or machine learning.

```python
import pandas as pd

# Create a DataFrame with full names
data = pd.DataFrame({
    "Full_Name": ["Arun Kumar", "Priya Sharma", "Kumar Raj"]
})

# Display the original data
print("Original Data:")
print(data)

# Split the full name into separate parts
# str.split() divides each string using the given separator.
data[["First_Name", "Last_Name"]] = data["Full_Name"].str.split(
    " ", n=1, expand=True
)

# Display the result
print("\nAfter using split():")
print(data)

# Expected output:
# Arun Kumar -> Arun, Kumar
# Priya Sharma -> Priya, Sharma
# Kumar Raj -> Kumar, Raj

# AI/ML / Data Analytics Use:
# split() helps extract useful information from text
# and create separate features for analysis or machine learning.
```

# contains()

## Understand the Concept

In Pandas, `contains()` is used to check whether a specific word, character, or pattern exists inside a string.

It returns `True` when the value contains the specified text and `False` when it does not.

### Why do we use it?

- To search for specific text.
- To filter rows based on text.
- To identify values containing a particular word.
- To perform text-based data cleaning.

### AI/ML Use

In AI/ML projects, `contains()` can be used to filter text data based on keywords.

For example, customer reviews containing words such as `good`, `bad`, or `delivery` can be identified for further analysis.

## Demonstrate the Concept

### Real-World Example

Suppose a company has customer feedback such as:

- "Good product"
- "Fast delivery"
- "Good quality"
- "Poor packaging"

We can use `contains()` to find all feedback that contains the word `"Good"`.

### AI/ML Example

Suppose we are analyzing customer reviews.

We can use `contains()` to identify reviews containing words such as `"good"` or `"bad"`.

These filtered reviews can then be used for sentiment analysis or other text-processing tasks.

```python
import pandas as pd

# Create a DataFrame containing customer feedback
data = pd.DataFrame({
    "Review": [
        "Good product",
        "Fast delivery",
        "Good quality",
        "Poor packaging"
    ]
})

# Display the original data
print("Original Data:")
print(data)

# Check whether each review contains the word "Good"
# str.contains() returns True or False for each row.
data["Contains_Good"] = data["Review"].str.contains(
    "Good",
    case=False
)

# Display the result
print("\nAfter using contains():")
print(data)

# Expected output:
# Reviews containing "Good" will have True.
# Other reviews will have False.

# AI/ML / Data Analytics Use:
# contains() can be used to filter text based on keywords
# before performing text analysis or machine learning.
```

# replace()

## Understand the Concept

In Pandas, `replace()` is used to replace existing values in a column with new values.

It can be used to replace specific words, characters, or values in text data.

### Why do we use it?

- To replace incorrect values.
- To clean text data.
- To standardize values.
- To correct inconsistent data.

### AI/ML Use

In AI/ML projects, datasets may contain inconsistent or incorrect text values.

For example, `M`, `Male`, and `male` may represent the same category.

We can use `replace()` to standardize these values before data analysis or machine learning.

## Demonstrate the Concept

### Real-World Example

Suppose a customer dataset contains the following gender values:

- Male
- M
- Female
- F

We can use `replace()` to convert `M` to `Male` and `F` to `Female`.

This makes the data more consistent.

### AI/ML Example

Suppose a machine learning dataset contains inconsistent category names.

For example, `"Yes"`, `"Y"`, and `"yes"` may represent the same response.

We can use `replace()` to convert them into one standard value before training the model.

```python
import pandas as pd

# Create a DataFrame with inconsistent gender values
data = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar", "Divya"],
    "Gender": ["M", "F", "M", "F"]
})

# Display the original data
print("Original Data:")
print(data)

# Replace short values with standard values
# replace() is used to change existing values.
data["Gender"] = data["Gender"].replace({
    "M": "Male",
    "F": "Female"
})

# Display the updated data
print("\nAfter using replace():")
print(data)

# Expected output:
# M -> Male
# F -> Female

# AI/ML / Data Analytics Use:
# replace() helps standardize inconsistent values
# before data analysis and machine learning.
```

# Regular Expressions (Regex)

## Understand the Concept

Regular Expressions, or Regex, are patterns used to search, identify, and manipulate text based on specific rules.

In Pandas, Regex can be used with string operations to find patterns inside text data.

### Why do we use it?

- To search for specific text patterns.
- To extract useful information from strings.
- To validate text formats.
- To clean complex text data.

### AI/ML Use

In AI/ML projects, Regex is useful for cleaning and extracting information from text data.

For example, we can use Regex to extract email addresses, phone numbers, or specific words from customer reviews and other text datasets.

## Demonstrate the Concept

### Real-World Example

Suppose a company has customer messages containing phone numbers.

For example:

`"Contact us at 9876543210"`

We can use a Regex pattern to find and extract the phone number from the text.

### AI/ML Example

Suppose we are processing customer reviews or messages.

Regex can be used to remove unwanted characters, extract email addresses or phone numbers, and clean the text before performing natural language processing or machine learning.

```python
import pandas as pd

# Create a DataFrame containing customer messages
data = pd.DataFrame({
    "Message": [
        "Contact us at 9876543210",
        "Call us at 8765432109",
        "Thank you for contacting us"
    ]
})

# Display the original messages
print("Original Data:")
print(data)

# Extract 10-digit phone numbers using Regex
# \d means a digit.
# {10} means exactly 10 digits.
data["Phone_Number"] = data["Message"].str.extract(
    r"(\d{10})"
)

# Display the extracted phone numbers
print("\nAfter using Regex:")
print(data)

# Expected output:
# The first two messages will have their 10-digit
# phone numbers extracted.
# The third message will contain NaN because
# no phone number is present.

# AI/ML / Data Analytics Use:
# Regex can be used to clean text and extract useful
# information such as phone numbers, email addresses,
# and other patterns before machine learning.
```
