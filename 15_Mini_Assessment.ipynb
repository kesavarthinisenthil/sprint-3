# Question 7: Pandas Data Manipulation

Create a Pandas DataFrame containing employee information with the columns:

- Name
- Age
- Department
- Salary

Perform the following operations:

1. Rename the `Salary` column to `Annual_Salary`.
2. Add a new column called `Bonus` using `assign()`.
3. Increase each salary by 10% using `apply()`.
4. Replace `IT` with `Technology` in the Department column.
5. Remove duplicate rows if any exist.

```python
import pandas as pd

# Create an employee DataFrame
employees = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar", "Divya", "Arun"],
    "Age": [25, 28, 30, 26, 25],
    "Department": ["IT", "HR", "Sales", "IT", "IT"],
    "Salary": [35000, 40000, 45000, 38000, 35000]
})

print("Original Data:")
print(employees)

# 1. Rename Salary column
employees = employees.rename(columns={"Salary": "Annual_Salary"})

# 2. Add Bonus column using assign()
employees = employees.assign(Bonus=employees["Annual_Salary"] * 0.10)

# 3. Increase salary by 10% using apply()
employees["Updated_Salary"] = employees["Annual_Salary"].apply(
    lambda salary: salary * 1.10
)

# 4. Replace IT with Technology
employees["Department"] = employees["Department"].replace(
    "IT", "Technology"
)

# 5. Remove duplicate rows
employees = employees.drop_duplicates()

# Display the final DataFrame
print("\nAfter Data Manipulation:")
print(employees)

# Expected output:
# Salary is renamed to Annual_Salary.
# Bonus is added.
# Updated_Salary contains a 10% increase.
# IT is replaced with Technology.
# Duplicate rows are removed.
```

# Question 8: GroupBy and Aggregation

Create a Pandas DataFrame containing employee information with the columns:

- Name
- Department
- Salary

Add at least 6 employees from different departments.

Perform the following operations:

1. Group employees by Department using `groupby()`.
2. Calculate the average salary for each department.
3. Calculate the total salary for each department using `agg()`.
4. Add the average department salary to each employee using `transform()`.

```python
import pandas as pd

# Create an employee DataFrame
employees = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar", "Divya", "Ravi", "Meena"],
    "Department": ["IT", "HR", "IT", "Sales", "HR", "Sales"],
    "Salary": [35000, 40000, 45000, 30000, 50000, 40000]
})

# Display the original data
print("Employee Data:")
print(employees)

# 1. Group employees by Department
grouped = employees.groupby("Department")

# 2. Calculate average salary for each department
print("\nAverage Salary by Department:")
print(grouped["Salary"].mean())

# 3. Calculate total salary using agg()
print("\nTotal Salary by Department:")
print(grouped["Salary"].agg("sum"))

# 4. Add department average salary to each employee
# transform() returns a value for every original row.
employees["Department_Average_Salary"] = (
    employees.groupby("Department")["Salary"].transform("mean")
)

# Display the updated DataFrame
print("\nEmployee Data with Department Average:")
print(employees)

# Expected output:
# Each department will have its average salary.
# Each employee will also have the average salary
# of their respective department.

# AI/ML / Data Analytics Use:
# groupby(), agg(), and transform() are useful for
# feature engineering and analyzing grouped data.
```

# Question 9: Merging DataFrames

Create two Pandas DataFrames:

1. A customer DataFrame containing:
   - Customer_ID
   - Name

2. A purchase DataFrame containing:
   - Customer_ID
   - Purchase_Amount

Use `merge()` to combine both DataFrames using `Customer_ID`.

Display the final DataFrame.

```python
import pandas as pd

# Create the customer DataFrame
customers = pd.DataFrame({
    "Customer_ID": [101, 102, 103, 104],
    "Name": ["Arun", "Priya", "Kumar", "Divya"]
})

# Create the purchase DataFrame
purchases = pd.DataFrame({
    "Customer_ID": [101, 102, 103, 104],
    "Purchase_Amount": [5000, 7000, 4500, 8000]
})

# Display the original DataFrames
print("Customer Data:")
print(customers)

print("\nPurchase Data:")
print(purchases)

# Merge both DataFrames using Customer_ID
# merge() combines related data using a common column.
combined = pd.merge(
    customers,
    purchases,
    on="Customer_ID"
)

# Display the merged DataFrame
print("\nMerged Data:")
print(combined)

# Expected output:
# The Name and Purchase_Amount columns will be combined
# using Customer_ID.

# AI/ML / Data Analytics Use:
# merge() is useful for combining related datasets
# before analysis, preprocessing, and feature engineering.
```

# Question 10: Date and Time Operations

Create a Pandas DataFrame containing an `Order_Date` column with at least 4 dates.

Perform the following operations:

1. Convert the `Order_Date` column to datetime using `pd.to_datetime()`.
2. Extract the year.
3. Extract the month.
4. Extract the day.
5. Extract the day name.
6. Calculate the number of days between the order date and a given date.

```python
import pandas as pd

# Create a DataFrame with order dates
orders = pd.DataFrame({
    "Order_ID": [101, 102, 103, 104],
    "Order_Date": [
        "2026-10-01",
        "2026-10-03",
        "2026-10-05",
        "2026-10-08"
    ]
})

# Convert Order_Date to datetime
orders["Order_Date"] = pd.to_datetime(orders["Order_Date"])

# Extract date features
orders["Year"] = orders["Order_Date"].dt.year
orders["Month"] = orders["Order_Date"].dt.month
orders["Day"] = orders["Order_Date"].dt.day
orders["Day_Name"] = orders["Order_Date"].dt.day_name()

# Calculate the number of days from the order date
# to October 10, 2026.
given_date = pd.Timestamp("2026-10-10")

orders["Days_Difference"] = (
    given_date - orders["Order_Date"]
).dt.days

# Display the final DataFrame
print(orders)

# Expected output:
# Year, Month, Day, and Day_Name will contain
# the extracted date information.
# Days_Difference will show the number of days
# between each order date and October 10, 2026.

# AI/ML / Data Analytics Use:
# Date features can be used for trend analysis,
# time-based predictions, and feature engineering.
```

# Question 11: String Operations

Create a Pandas DataFrame containing a `Name` column and a `Message` column.

Perform the following string operations:

1. Convert names to lowercase using `lower()`.
2. Convert names to uppercase using `upper()`.
3. Remove extra spaces using `strip()`.
4. Check whether messages contain the word `"good"` using `contains()`.
5. Replace a word in the messages using `replace()`.

```python
import pandas as pd

# Create a DataFrame containing text data
data = pd.DataFrame({
    "Name": [" Arun ", " PRIYA ", " Kumar "],
    "Message": [
        "Good product",
        "Good service",
        "Bad product"
    ]
})

# Remove extra spaces from names
data["Clean_Name"] = data["Name"].str.strip()

# Convert names to lowercase
data["Lower_Name"] = data["Clean_Name"].str.lower()

# Convert names to uppercase
data["Upper_Name"] = data["Clean_Name"].str.upper()

# Check whether the message contains the word "good"
# case=False makes the search case-insensitive.
data["Contains_Good"] = data["Message"].str.contains(
    "good",
    case=False
)

# Replace "Bad" with "Needs Improvement"
data["Updated_Message"] = data["Message"].str.replace(
    "Bad",
    "Needs Improvement",
    regex=False
)

# Display the final DataFrame
print(data)

# Expected output:
# Clean_Name removes extra spaces.
# Lower_Name contains lowercase names.
# Upper_Name contains uppercase names.
# Contains_Good shows True or False.
# Updated_Message replaces "Bad" with "Needs Improvement".

# AI/ML / Data Analytics Use:
# String operations are useful for cleaning text data
# before analysis, NLP, and machine learning.
```

# Question 12: Missing Values

Create a Pandas DataFrame containing:

- Name
- Age
- City

Include some missing values in the `Age` and `City` columns.

Perform the following operations:

1. Check for missing values using `isnull()`.
2. Count the missing values in each column.
3. Fill missing `Age` values using the mean.
4. Fill missing `City` values using the mode.
5. Display the cleaned DataFrame.

```python
import pandas as pd

# Create a DataFrame with missing values
data = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar", "Divya", "Ravi"],
    "Age": [25, None, 30, 28, None],
    "City": ["Chennai", "Salem", None, "Coimbatore", "Chennai"]
})

# 1. Check for missing values
print("Missing Values:")
print(data.isnull())

# 2. Count missing values in each column
print("\nMissing Value Count:")
print(data.isnull().sum())

# 3. Fill missing Age values using the mean
mean_age = data["Age"].mean()
data["Age"] = data["Age"].fillna(mean_age)

# 4. Fill missing City values using the mode
mode_city = data["City"].mode()[0]
data["City"] = data["City"].fillna(mode_city)

# 5. Display the cleaned DataFrame
print("\nCleaned Data:")
print(data)

# Expected output:
# Missing Age values are replaced with the mean age.
# Missing City values are replaced with the most common city.

# AI/ML / Data Analytics Use:
# Handling missing values is an important preprocessing
# step before training machine learning models.
```

# Question 13: Performance Optimization

Create a Pandas DataFrame containing:

- Name
- Age
- Department
- Salary

Perform the following operations:

1. Use vectorization to increase all salaries by 10%.
2. Use `apply()` to create a salary category.
3. Convert the `Department` column to `category` dtype.
4. Display the data types after optimization.

```python
import pandas as pd

# Create an employee DataFrame
employees = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar", "Divya", "Ravi"],
    "Age": [25, 28, 30, 26, 32],
    "Department": ["IT", "HR", "IT", "Sales", "IT"],
    "Salary": [35000, 60000, 45000, 30000, 70000]
})

# 1. Use vectorization to increase salaries by 10%
# The entire column is processed at once.
employees["Updated_Salary"] = employees["Salary"] * 1.10

# 2. Create a salary category using apply()
def salary_category(salary):
    if salary >= 50000:
        return "High Salary"
    else:
        return "Normal Salary"

employees["Salary_Category"] = employees["Salary"].apply(
    salary_category
)

# 3. Convert Department to category dtype
employees["Department"] = employees["Department"].astype("category")

# Display the final DataFrame
print("Optimized Data:")
print(employees)

# 4. Display data types
print("\nData Types:")
print(employees.dtypes)

# Expected output:
# Updated_Salary contains a 10% increase.
# Salary_Category contains High Salary or Normal Salary.
# Department has category dtype.

# AI/ML / Data Analytics Use:
# Vectorization, suitable data types, and efficient
# transformations help improve performance when working
# with large datasets.
```

# Question 14: Combined NumPy + Pandas Assessment

Create a Pandas DataFrame containing the following student information:

- Name
- Math
- Science
- English

Add at least 5 students.

Perform the following operations:

1. Use NumPy to calculate the average marks for each student.
2. Add the average marks as a new column called `Average`.
3. Create a `Result` column using `apply()`:
   - If Average is 50 or above, assign `"Pass"`.
   - Otherwise, assign `"Fail"`.
4. Find the student with the highest average.
5. Sort the DataFrame by Average in descending order.

```python
import pandas as pd
import numpy as np

# Create a student DataFrame
students = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar", "Divya", "Ravi"],
    "Math": [80, 65, 45, 90, 55],
    "Science": [75, 70, 40, 85, 60],
    "English": [85, 60, 50, 95, 65]
})

# Display the original data
print("Student Data:")
print(students)

# 1. Calculate average marks using NumPy
# np.mean() calculates the average of the three subject marks.
students["Average"] = np.mean(
    students[["Math", "Science", "English"]],
    axis=1
)

# 2. Create Result column using apply()
# Students with an average of 50 or above are marked as Pass.
def calculate_result(average):
    if average >= 50:
        return "Pass"
    else:
        return "Fail"

students["Result"] = students["Average"].apply(
    calculate_result
)

# 3. Find the student with the highest average
highest_student = students.loc[
    students["Average"].idxmax()
]

print("\nStudent with Highest Average:")
print(highest_student)

# 4. Sort students by Average in descending order
students = students.sort_values(
    "Average",
    ascending=False
)

# Display the final DataFrame
print("\nFinal Student Data:")
print(students)

# Expected output:
# Average contains the average marks of each student.
# Result contains Pass or Fail.
# The student with the highest average is displayed.
# The final DataFrame is sorted from highest to lowest average.

# AI/ML / Data Analytics Use:
# NumPy can perform efficient numerical calculations,
# while Pandas can organize, transform, and analyze
# the resulting data.
```
